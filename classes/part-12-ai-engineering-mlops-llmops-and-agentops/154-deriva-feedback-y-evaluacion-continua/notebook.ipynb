## 📖 Materia resumida

**Tres derivas**: de datos (`P(X)` cambia; alarma temprana, no veredicto), de etiquetas
(`P(Y)` cambia; recalibrar umbral), de concepto (`P(Y|X)` cambia; la más grave — solo las
etiquetas la confirman).

**PSI** sobre B bins con baseline `p` y actual `q`:

```text
PSI = Σ (q_i − p_i) · ln(q_i / p_i)   ≥ 0
< 0.10 estable · 0.10–0.25 investigar · > 0.25 actuar   (convención, no teorema)
```

**Bucle de feedback**: proxies inmediatos (distribución del score, fallbacks) →
etiquetas diferidas por cohortes (retardo y sesgo de selección) → muestreo etiquetado
continuo → política de acción escrita (qué umbral dispara qué). Reentrenar solo cura la
deriva causada por el mundo; si la causa es un bug de datos, reentrenar lo aprende.
