## 📖 Materia resumida

**Experimento** = `f(código, datos, configuración, azar) → métricas + artefactos`.
El tracking registra todas las entradas y salidas: `params` (para relanzar), `metrics`
(para decidir), `artifacts` (para desplegar), `tags` (commit, autor).

**Fuentes de no determinismo**: semillas (controlables gratis), paralelismo en punto
flotante (controlable con costo), entorno (lockfiles/contenedores), datos vivos
(snapshots con hash).

**Linaje**: grafo dirigido `datos → features → run → modelo`, con hash de contenido por
nodo. Hacia atrás responde procedencia; hacia adelante, análisis de impacto.

**Protocolo honesto**: semilla fijada antes de mirar resultados; ≥ 3 semillas al comparar
métodos; reportar media ± desviación, nunca el mejor run.


## 🧮 Ejemplo de referencia

```text
config A: AUC = 0.812, 0.804, 0.808  →  media 0.808, desv ≈ 0.004
config B: AUC = 0.815, 0.799, 0.802  →  media 0.805, desv ≈ 0.009
```

El mejor run individual es de B (0.815), pero la media de A es superior y la diferencia
de medias (0.003) es menor que la variación entre semillas de B (±0.009): no hay
evidencia para elegir. Elegir B por su mejor run sería *seed picking* — sobreajuste al
azar de la semilla.
