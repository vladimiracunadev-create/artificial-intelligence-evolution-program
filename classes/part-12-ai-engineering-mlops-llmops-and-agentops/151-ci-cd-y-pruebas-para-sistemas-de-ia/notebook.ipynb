## 📖 Materia resumida

En IA, un cambio puede venir de **código, datos o modelo**, y cada eje tiene sus tests:

- **Datos**: esquema (tipos, rangos, nulos) bloquea; distribución (medias, cuantiles)
  alerta.
- **Modelo**: calidad vs. baseline con umbral; paridad offline/online (skew);
  reproducibilidad (semilla + snapshot).
- **Comportamiento**: invariancia (cambiar un nombre propio no cambia el sentimiento),
  direccionales (más ingreso ⇒ no más impago), casos mínimos que jamás fallan.

**CI** valida código+datos y produce un pipeline probado; **CD** despliega el pipeline;
**CT** (continuous training) reentrena por calendario/datos/deriva — la pieza que no
existe en DevOps clásico. Regla de diseño: los contratos **bloquean**, las señales
estadísticas ruidosas **alertan**.
