## 📖 Materia resumida

Un sistema de IA operado encadena **tres ciclos de vida** con ritmos distintos:

- **Datos**: ingesta → validación → versionado. Artefacto: dataset con esquema,
  estadísticas y hash. Cambia al ritmo del mundo.
- **Modelo**: entrenamiento → evaluación → registro → despliegue → monitoreo.
  Artefacto: modelo registrado (pesos + código + entorno + métricas). Cambia por decisión.
- **Agente**: prompt + herramientas → evals de trayectoria → despliegue con límites.
  Artefacto: configuración versionada. Cambia con cada ajuste — y también cuando el
  proveedor cambia el modelo base.

Regla de oro: cada transición produce un artefacto con **identidad** (con qué datos,
código y semilla se produjo). Sin linaje no hay diagnóstico posible.


## 🧮 Ejemplo de referencia

El diagnóstico recorre los ciclos por velocidad de cambio: datos → modelo → agente.
Aquí la causa vive en el ciclo de datos (el tracking de la app cambió el viernes);
el modelo no cambió. La conclusión solo es accionable porque cada artefacto tiene id y
estadísticas registradas.
