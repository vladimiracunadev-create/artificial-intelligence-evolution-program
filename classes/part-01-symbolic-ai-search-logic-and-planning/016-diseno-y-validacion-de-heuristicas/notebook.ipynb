## 📖 Resumen de la materia

Una **heurística** `h(n)` estima el costo restante hasta el objetivo. Su diseño
gobierna el rendimiento de A*:

- **Admisible**: `h(n) ≤ h*(n)` para todo `n` (nunca sobreestima) → A* es óptimo.
- **Consistente**: `h(n) ≤ c(n, n') + h(n')` (desigualdad triangular) → A* no
  necesita reabrir nodos ya expandidos. Consistente ⇒ admisible.
- **Dominancia**: si `h2(n) ≥ h1(n)` para todo `n` (ambas admisibles), `h2`
  domina a `h1` y A* con `h2` expande un subconjunto de los nodos.
- **Receta estándar**: resolver de forma exacta un **problema relajado**
  (quitar restricciones) produce heurísticas admisibles automáticamente.
  También: `max(h1, h2)` de admisibles es admisible y domina a ambas.


## 🧮 Las dos heurísticas clásicas del 8-puzzle

- `h1` = **fichas mal colocadas** (relajación: una ficha "salta" a su sitio).
- `h2` = **distancia Manhattan** = suma de `|Δfila| + |Δcolumna|` de cada ficha
  (relajación: las fichas se deslizan atravesándose).

`h2 ≥ h1` siempre (cada ficha mal colocada aporta ≥ 1 a Manhattan), así que
`h2` domina a `h1`. Empíricamente (AIMA): a profundidad 12, A* con `h1` expande
un orden de magnitud más nodos que con `h2`. La calidad se mide con el
**factor de ramificación efectivo** `b*`: el `b` de un árbol uniforme que
generaría los mismos nodos — cuanto más cerca de 1, mejor la heurística.
