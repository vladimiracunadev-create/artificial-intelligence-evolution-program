## 📖 Resumen de la materia

Un **CSP** se define con variables `X`, dominios `D` y restricciones `C`; una
solución es una asignación completa y consistente. A diferencia de la búsqueda
atómica, el estado está **factorizado** y eso permite dos palancas:

- **Backtracking con heurísticas generales**: MRV (elegir la variable con menos
  valores legales), grado (la más restrictiva) y valor menos restrictivo.
- **Propagación de restricciones**: eliminar valores imposibles *antes* de
  probarlos. **AC-3** impone consistencia de arco: `X→Y` es consistente si todo
  valor de `X` tiene algún compatible en `Y`; se revisan arcos en cola hasta el
  punto fijo. Costo `O(cd³)`. Forward checking es la versión ligera tras cada
  asignación.

El laboratorio de esta clase usa el motor `logic` (encadenamiento de reglas):
propagar restricciones y encadenar reglas comparten la misma idea — derivar
consecuencias forzadas sin buscar.


## 🧮 Mini-CSP de referencia

`X, Y, Z ∈ {1, 2, 3}` con restricciones `X < Y` y `Y < Z`.

AC-3 reduce dominios sin buscar: de `X<Y`, X pierde el 3 e Y pierde el 1; de
`Y<Z`, Y pierde el 3 y Z pierde el 1; re-revisando `X<Y` con Y={2}, X={1}; y
`Y<Z` con Z={3}. Dominios finales: `X={1}, Y={2}, Z={3}` — la propagación sola
resolvió el problema. En CSPs reales (mapa de Australia con 3 colores,
sudoku, horarios) la propagación no basta y se intercala con backtracking.
