## 📖 Resumen de la materia

Cuando las acciones tienen costos distintos, contar aristas (BFS) deja de ser
óptimo. Las tres estrategias con **cola de prioridad** se distinguen por la
función de evaluación `f(n)` con la que ordenan la frontera:

| Algoritmo | f(n) | Óptimo | Usa |
|---|---|---|---|
| Costo uniforme (UCS) | `g(n)` (costo acumulado) | sí | solo el pasado |
| Voraz (greedy) | `h(n)` (heurística) | **no** | solo la estimación |
| A* | `g(n) + h(n)` | sí, si `h` es admisible | pasado + futuro estimado |

- `h(n)` **admisible**: nunca sobreestima el costo real restante (`h ≤ h*`).
- `h(n)` **consistente**: `h(n) ≤ c(n, n') + h(n')` — garantiza que A* puede
  dar por definitivo un nodo al expandirlo (versión en grafo).
- UCS es A* con `h = 0`; Dijkstra es UCS a todos los destinos.
