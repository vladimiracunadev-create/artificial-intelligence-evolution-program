## 📖 Resumen de la materia

Representar conocimiento es decidir **qué existe y cómo se relaciona** para que
un programa pueda razonar sobre ello:

- **Ontología**: el vocabulario formal de un dominio — clases (categorías),
  individuos, relaciones y axiomas. Distinciones que no se pueden confundir:
  - `instance-of` (Fido **es un** individuo de la clase Perro),
  - `is-a` / subclase (Perro **es una subclase de** Mamífero),
  - `part-of` (Rueda es **parte de** Coche — no hereda propiedades).
- **Herencia**: las propiedades de una clase se transmiten a subclases e
  instancias; con **excepciones** (los pingüinos no vuelan) la lógica clásica
  se vuelve no monótona: hay que usar valores por defecto anulables.
- **Tripletas (RDF)**: todo se expresa como `(sujeto, predicado, objeto)`;
  los grafos de conocimiento (Wikidata, Google KG) escalan este modelo.
  OWL añade axiomas (disjunción, cardinalidad, transitividad) con semántica
  de lógica de descripciones, donde la inferencia es decidible.


## 🧮 Micro-ontología de referencia

```text
(Pinguino,  subclase_de,  Ave)        (Ave,   subclase_de, Animal)
(Ave,       vuela,        true)       (Pinguino, vuela,    false)   ← excepción
(Pingu,     instancia_de, Pinguino)   (Ala,   parte_de,    Ave)
```

Consulta "¿Pingu vuela?": la herencia simple sube `Pingu → Pinguino → Ave` y
respondería `true`; con **la excepción más específica primero**, responde
`false`. Y "¿Pingu tiene alas?" requiere distinguir: hereda *tener* alas vía
`is-a`, pero `parte_de` no se hereda hacia las partes (el ala no es un ave).
El laboratorio `logic` muestra la maquinaria mínima: hechos + reglas + cierre.
