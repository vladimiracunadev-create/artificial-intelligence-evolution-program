## 📖 Resumen de la materia

La lógica proposicional da a un agente un lenguaje con semántica precisa:

- **Implicación (⊨)**: `KB ⊨ α` si α es verdadera en **todo** modelo donde KB
  lo es. Verificable por tabla de verdad: `2^n` filas para n símbolos.
- **Inferencia correcta (sound)**: solo deriva cosas implicadas. **Completa**:
  deriva todo lo implicado.
- **Resolución**: única regla, opera sobre **CNF** (conjunción de cláusulas).
  Para probar `KB ⊨ α` se añade `¬α` y se buscan resolventes hasta derivar la
  cláusula vacía `□` (refutación). Correcta y completa para proposicional.
- **Cláusulas de Horn** (≤ 1 literal positivo): permiten **encadenamiento
  hacia adelante/atrás** en tiempo lineal — el mecanismo exacto del laboratorio
  `logic`, donde cada regla `p1 ∧ p2 → q` dispara cuando sus premisas están en
  la memoria de hechos.


## 🧮 Conversión a CNF en 4 pasos

```text
1. Eliminar ↔:   α ↔ β  ⇒  (α → β) ∧ (β → α)
2. Eliminar →:   α → β  ⇒  ¬α ∨ β
3. Empujar ¬ hacia dentro (De Morgan, doble negación)
4. Distribuir ∨ sobre ∧
```

Ejemplo: `(P → Q) → R` ⇒ `¬(¬P ∨ Q) ∨ R` ⇒ `(P ∧ ¬Q) ∨ R` ⇒
`(P ∨ R) ∧ (¬Q ∨ R)`. Dos cláusulas listas para resolución.
