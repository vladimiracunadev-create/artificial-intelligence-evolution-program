## 📖 Resumen de la materia

Un sistema **neuro-simbólico** reparte el trabajo: percepción y propuesta al
componente aprendido; verificación, razonamiento y **explicación fiel** al
componente simbólico. La **taxonomía de Kautz** (AAAI 2020 / AI Magazine 2022)
da nombre a las arquitecturas:

```text
Tipo 1  symbolic Neuro symbolic    (un LLM de texto a texto, visto como caja)
Tipo 2  Symbolic[Neuro]            (AlphaGo: búsqueda simbólica llama a la red)
Tipo 3  Neuro; Symbolic            (pipeline: red extrae símbolos → razonador)
Tipo 4  Neuro: Symbolic → Neuro    (conocimiento compilado en el entrenamiento)
Tipo 5  Neuro_{Symbolic}           (reglas tensorizadas dentro de la red)
Tipo 6  Neuro[Symbolic]            (razonamiento combinatorio embebido; aspiracional)
```

Distinción clave del proyecto: **explicación fiel** (la traza de reglas ES el
cómputo) vs. **post-hoc** (racionalización sobre un modelo opaco, plausible
pero no necesariamente verdadera).
