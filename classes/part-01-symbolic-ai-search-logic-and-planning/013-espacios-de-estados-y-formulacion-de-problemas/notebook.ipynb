## 📖 Resumen de la materia

Un **problema de búsqueda** bien formulado tiene cinco componentes (AIMA 4e, cap. 3):

1. **Estado inicial** `s0`.
2. **Acciones** `ACTIONS(s)`: acciones aplicables en `s`.
3. **Modelo de transición** `RESULT(s, a)`: estado resultante.
4. **Test de objetivo** `IS-GOAL(s)`: predicado (puede haber muchos estados meta).
5. **Costo** `c(s, a, s') ≥ 0`; el costo de un camino es la suma de sus pasos.

El **espacio de estados** es el grafo dirigido *implícito* de todos los estados
alcanzables: casi nunca se materializa, se genera bajo demanda con `ACTIONS` y
`RESULT`. Distinción clave: un **estado** es una configuración del mundo; un
**nodo** de búsqueda añade padre, acción y costo acumulado `g(n)` — dos nodos
distintos pueden contener el mismo estado.
