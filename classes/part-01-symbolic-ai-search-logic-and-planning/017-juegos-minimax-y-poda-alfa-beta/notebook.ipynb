## 📖 Resumen de la materia

En juegos de suma cero por turnos, **minimax** define el valor de un estado
asumiendo juego perfecto de ambos: MAX elige el hijo de valor máximo, MIN el de
valor mínimo, y las hojas valen su utilidad.

```text
MINIMAX(s) = utilidad(s)                    si s es terminal
           = max hijos MINIMAX              si mueve MAX
           = min hijos MINIMAX              si mueve MIN
```

La **poda alfa-beta** calcula el mismo valor sin explorar ramas irrelevantes:
`α` = mejor valor garantizado para MAX en el camino; `β` = mejor para MIN.
Cuando en un nodo MIN el valor cae a `≤ α`, MAX jamás entrará ahí: se **poda**
el resto de hijos (y simétricamente `≥ β` en nodos MAX). Con **ordenación
perfecta** de movimientos, alfa-beta examina `O(b^(d/2))` hojas — duplica la
profundidad alcanzable con el mismo presupuesto.

```text

```

Minimax: `B = 3`, `C = 2`, `D = 2` → raíz = **3** (mover a B).
En los ejercicios se ejecuta alfa-beta de izquierda a derecha sobre este árbol

En profundidad práctica los valores de hoja los da una **función de evaluación**
truncada, no la utilidad real — ahí viven los errores del jugador artificial.
