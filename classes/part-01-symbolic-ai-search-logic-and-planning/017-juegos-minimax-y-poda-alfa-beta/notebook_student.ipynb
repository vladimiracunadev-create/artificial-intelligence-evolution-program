```python
# TU RESPUESTA AQUÍ (ejercicio 3) — opcional: verifica con código
def alfabeta(nodo, es_max, alfa, beta, contador):
    if isinstance(nodo, (int, float)):
        contador[0] += 1
        return nodo
    if es_max:
        v = float("-inf")
        for hijo in nodo:
            v = max(v, alfabeta(hijo, False, alfa, beta, contador))
            alfa = max(alfa, v)
            if alfa >= beta:
                break
        return v
    v = float("inf")
    for hijo in nodo:
        v = min(v, alfabeta(hijo, True, alfa, beta, contador))
        beta = min(beta, v)
        if alfa >= beta:
            break
    return v

arbol = [[3, 12, 8], [2, 4, 6], [14, 5, 2]]
# Prueba distintos órdenes de los tres hijos y cuenta hojas evaluadas.
```
