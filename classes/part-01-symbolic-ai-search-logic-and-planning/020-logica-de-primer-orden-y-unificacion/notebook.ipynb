## 📖 Resumen de la materia

La **lógica de primer orden (FOL)** añade objetos, relaciones, funciones y
cuantificadores. Convenciones de traducción que hay que dominar:

- `∀x (Estudiante(x) → Aprueba(x))` — "todos los estudiantes aprueban":
  ∀ va con **→** (con ∧ diría "todo es estudiante y aprueba").
- `∃x (Estudiante(x) ∧ Aprueba(x))` — "algún estudiante aprueba":
  ∃ va con **∧** (con → sería casi vacuamente verdadera).

La **unificación** es el motor de la inferencia FOL: `UNIFICA(p, q)` devuelve
la sustitución más general (**MGU**) que hace `p = q`, o falla.

```text
UNIFICA(Conoce(Juan, x), Conoce(y, Madre(y))) = {y/Juan, x/Madre(Juan)}
```

Reglas: variable contra término liga (salvo **occurs check**: `x` no puede
unificar con `f(x)`); constantes distintas fallan; funciones/predicados deben
coincidir en nombre y aridad. El **modus ponens generalizado** aplica reglas
con variables unificando sus premisas contra hechos — es el encadenamiento del
laboratorio elevado a primer orden.


## 🧮 Por qué la MGU debe ser "la más general"

`Conoce(Juan, x)` y `Conoce(y, z)` unifican con infinitas sustituciones:
`{y/Juan, x/z}`, `{y/Juan, x/Ana, z/Ana}`, … La MGU es `{y/Juan, x/z}`: no
compromete nada innecesario, y cualquier otra unificación es una instancia de
ella. Elegir una sustitución menos general puede bloquear pruebas posteriores
— por eso los algoritmos (y Prolog) computan siempre la MGU. Antes de unificar
dos cláusulas se **estandarizan aparte** las variables (renombrar `x` de una y
`x` de otra) para evitar capturas accidentales.
