## 📖 Resumen de la materia

**BFS** (anchura) y **DFS** (profundidad) son el mismo esqueleto de búsqueda con
distinta estructura de frontera:

- **BFS** usa una **cola FIFO**: expande por niveles. Es **completa** y devuelve
  el camino con menos aristas (óptimo si todos los costos son iguales).
  Tiempo y memoria: `O(b^d)` — la memoria es su límite práctico.
- **DFS** usa una **pila LIFO**: baja por una rama hasta agotarla. Memoria
  `O(b·m)` (m = profundidad máxima), pero **no es completa** en espacios
  infinitos ni óptima: devuelve el primer camino que encuentra.
- **Búsqueda en profundidad iterativa (IDS)**: DFS con límite creciente;
  combina la memoria de DFS con la completitud/optimalidad de BFS a costo
  de re-expandir niveles superiores (sobrecosto pequeño porque `b^d` domina).


## 🧮 Traza de referencia (grafo del laboratorio)

Grafo: `A→B,C; B→D,E; C→F; E→G; F→G`, inicial `A`, objetivo `G`.

```text

DFS (apilando [B,C] y sacando el último):
     expande A, C, F, G            → camino A→C→F→G (3 aristas)
```
