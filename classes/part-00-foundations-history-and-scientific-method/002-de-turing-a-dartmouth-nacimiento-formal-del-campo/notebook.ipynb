## 📖 Resumen de la materia

Entre 1936 y 1956 se ensamblan las piezas fundacionales del campo:

- **1936 — Turing, *On Computable Numbers*:** define la máquina universal (fundamento del
  software) y demuestra la indecidibilidad del Entscheidungsproblem: la IA nace dentro de
  límites formales conocidos.
- **1943 — McCulloch y Pitts:** la neurona formal binaria con umbral; redes de estas
  unidades computan cualquier función booleana. Ancestro directo de las redes actuales.
- **1950 — Turing, *Computing Machinery and Intelligence* (Mind, LIX-236):** reemplaza
  "¿pueden pensar las máquinas?" por el **juego de imitación**, un criterio operativo, y
  propone máquinas que aprenden como camino práctico.
- **1956 — Dartmouth:** McCarthy, Minsky, Rochester y Shannon acuñan *artificial
  intelligence* y formulan la conjetura fundacional: todo aspecto de la inteligencia puede
  describirse con precisión suficiente para simularse.
- **1959 — Samuel:** su programa de damas aprende de la experiencia y supera a su autor.
