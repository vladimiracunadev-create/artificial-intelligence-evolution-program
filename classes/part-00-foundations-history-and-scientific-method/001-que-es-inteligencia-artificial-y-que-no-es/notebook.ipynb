## 📖 Resumen de la materia

**Definición operativa (AIMA 4e, cap. 1):** la IA moderna se define por el **agente
racional** — un sistema que percibe su entorno y actúa para maximizar el valor esperado de
una medida de desempeño. No exige conciencia: exige desempeño medible.

Distinciones clave de esta clase:

- **IA estrecha:** competente en una tarea acotada (clasificar, traducir, jugar). Todo
  sistema desplegado hoy es IA estrecha; su competencia no transfiere fuera de su
  distribución de datos.
- **IA general (AGI):** hipotética competencia transferible a la mayoría de dominios.
  Es un objetivo de investigación, no un artefacto existente.
- **Automatización:** ejecutar un procedimiento *especificado de antemano* (una macro, un
  cron job). No es IA aunque no intervenga un humano.
- **Autonomía:** seleccionar acciones ante situaciones *no enumeradas* por el diseñador,
  siempre acotada por un dominio de operación (ODD).
