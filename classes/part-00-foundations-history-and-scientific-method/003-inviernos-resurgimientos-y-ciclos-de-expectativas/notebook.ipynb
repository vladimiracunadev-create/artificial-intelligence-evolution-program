## 📖 Resumen de la materia

Un **invierno de IA** es un colapso de financiación y credibilidad tras expectativas
infladas. El patrón histórico tiene seis pasos: avance real acotado → extrapolación pública
→ financiación atada a la promesa → muro técnico → informe/fracaso que cristaliza →
recorte y éxodo.

- **Primer invierno (~1974-1980):** detonado por el informe ALPAC (1966, traducción
  automática), *Perceptrons* de Minsky y Papert (1969, límite XOR) y el **informe
  Lighthill** (1973), que identificó la **explosión combinatoria** como muro de los métodos
  de búsqueda general.
- **Segundo invierno (~1987-1993):** los sistemas expertos (XCON, MYCIN) dieron valor real
  pero colapsaron por costo de mantenimiento de reglas frágiles y el desplome del mercado
  de máquinas Lisp.
- **Resurgimiento (1995→):** métodos probabilísticos, evaluación en benchmarks y, desde
  2012 (AlexNet), deep learning con GPU y datos masivos.

Indicadores para separar avance real de burbuja: mejora en benchmarks independientes y
fuera de distribución, costo por tarea descendente, despliegues reales con usuarios que
pagan, y promesas con fecha y criterio falsable.
