## 📖 Resumen de la materia

- **Leyes de potencia**: la pérdida de preentrenamiento cae de forma suave y
  predecible con parámetros N, tokens D y cómputo C (Kaplan et al., 2020).
- **Presupuesto**: C ≈ 6·N·D FLOPs; elegir modelo = repartir C entre N y D.
- **Chinchilla**: L(N,D) = E + A/N^α + B/D^β; el óptimo es N* ∝ C^0,5 y D* ∝ C^0,5,
  regla práctica ~20 tokens/parámetro. Chinchilla 70B/1,4T > Gopher 280B/300B con
  el mismo cómputo.
- **Matiz operativo**: el óptimo de entrenamiento no es el de inferencia; para
  servir barato conviene sobre-entrenar modelos pequeños (Llama 3 8B: >1 800
  tokens/parámetro).
- **Emergencia**: la pérdida es predecible; las capacidades por tarea, no siempre
  (y parte de la "emergencia" es artefacto de métricas discontinuas).
