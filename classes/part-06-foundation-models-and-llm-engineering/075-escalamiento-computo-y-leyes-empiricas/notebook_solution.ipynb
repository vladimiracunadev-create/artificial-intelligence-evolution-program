## Solución de referencia

**Ejercicio 1.** C = 120·N² → N* = sqrt(2,4·10²²/120) = sqrt(2·10²⁰) ≈ 1,4·10¹⁰
(**~14B parámetros**); D* = 20·N* ≈ 2,8·10¹¹ (**~280B tokens**).

**Ejercicio 2.** 200B/40B = **5 tokens/parámetro** → muy por debajo de ~20:
**sub-entrenado** (con ese cómputo, Chinchilla sugeriría ~11B parámetros y ~230B
tokens). C = 6·4·10¹⁰·2·10¹¹ = **4,8·10²² FLOPs**.

**Ejercicio 3.** L(10¹⁰) = (5,4·10¹³/10¹⁰)^0,095 = 5400^0,095 ≈ e^(0,095·8,594) ≈
**2,26**; L(10¹²) = 54^0,095 ≈ **1,46**. Para bajar 10 %:
k = 0,9^(−1/0,095) ≈ e^(0,105/0,095·...) ≈ **~3**: hay que ~triplicar los datos
por cada −10 % de pérdida. Las leyes de potencia son implacables.

**Ejercicio 4.** El laboratorio compara y evalúa resultados bajo contrato
reproducible; no entrena nada — por eso es `evaluation`.

```python
import math

# Ejercicio 1
C = 2.4e22
N_opt = math.sqrt(C / 120)
D_opt = 20 * N_opt
print(f"N*={N_opt:.2e}  D*={D_opt:.2e}")  # ~1.4e10, ~2.8e11

# Ejercicio 2
N, D = 40e9, 200e9
print("tokens/param =", D / N, " C =", f"{6*N*D:.2e}")  # 5.0, 4.8e22

# Ejercicio 3
L = lambda D_: (5.4e13 / D_) ** 0.095
k = 0.9 ** (-1 / 0.095)
print(f"L(1e10)={L(1e10):.2f}  L(1e12)={L(1e12):.2f}  k={k:.2f}")

# Ejercicio 4
result = run_lab("evaluation", seed=75)
assert result["kind"] == "evaluation"
assert result["evidence"] and result["limitations"]
show(result)
```
