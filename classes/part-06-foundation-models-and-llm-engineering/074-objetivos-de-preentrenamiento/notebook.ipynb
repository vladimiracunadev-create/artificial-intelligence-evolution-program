## 📖 Resumen de la materia

- **Auto-supervisión**: la etiqueta sale del propio texto; el modelo minimiza
  entropía cruzada sobre billones de tokens y aprende P(texto).
- **Causal / next-token (GPT)**: P(x) = Π P(xᵢ|x<ᵢ), decoder-only con máscara
  causal; 100 % de posiciones dan señal y el modelo genera texto nativo.
- **MLM (BERT)**: enmascara ~15 % y predice esos tokens con atención bidireccional;
  excelente para comprensión, incapaz de generar.
- **Span denoising (T5)**: reconstruye spans eliminados con encoder-decoder y
  formula todo como texto-a-texto.
- **Métrica**: pérdida media por token y perplejidad = e^L; no "accuracy".

Ejemplo: si el modelo asigna 0,20/0,25/0,10 al token correcto en tres posiciones,
L media ≈ 1,766 y perplejidad ≈ 5,85 ("duda entre ~6 opciones por token").
