## Solución de referencia

**Ejercicio 1.** Un buen par: un caso claro de `facturacion` ("me cobraron dos
veces") y un caso **frontera** resuelto ("no puedo pagar porque la página da error
500" → `tecnico`, porque el bloqueo es técnico aunque hable de pago). El ejemplo
frontera enseña el criterio de desempate, que es lo que zero-shot no sabe.

**Ejercicio 2.** (1) Texto extra alrededor del JSON → prefill/plantilla que arranca
en `{`. (2) JSON malformado (coma final, comillas) → validación + reintento o
decodificación restringida. (3) Campos faltantes o con tipo incorrecto → JSON
Schema con `required` y `type`. (Bonus: dato plausible pero falso → validación
semántica de negocio; ninguna capa sintáctica lo atrapa.)

**Ejercicio 3.** Defectuosos = 240·0,15 = 36. Reparables = 36·0,40 = 14,4 → según
el enunciado real serían 14 o el dato es continuo; con redondeo entero: 14.
Irreparables = 36 − 14 = **22** (o 21,6 sin redondear). La cadena explícita expone
además la ambigüedad del redondeo — valor pedagógico del CoT.

**Ejercicio 4.** El laboratorio devuelve un contrato JSON con claves obligatorias:
es exactamente el patrón "salida estructurada validable" aplicado a la evidencia
educativa.

```python
# Ejercicio 3 verificado
defectuosos = 240 * 0.15          # 36
reparables = defectuosos * 0.40   # 14.4
irreparables = defectuosos - reparables
print(defectuosos, reparables, irreparables)  # 36.0 14.4 21.6

# Ejercicio 4
result = run_lab("llm", seed=79)
assert result["kind"] == "llm"
assert result["evidence"] and result["limitations"]
show(result)
```
