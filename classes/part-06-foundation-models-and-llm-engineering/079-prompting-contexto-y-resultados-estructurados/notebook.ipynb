## 📖 Resumen de la materia

- **In-context learning**: zero-shot (solo instrucción) vs few-shot (k ejemplos
  resueltos que fijan formato y criterio sin tocar pesos). El orden y formato de
  los ejemplos importan; hay sesgo de recencia.
- **Chain-of-thought**: pedir pasos intermedios convierte la salida del modelo en
  memoria de trabajo y desbloquea tareas multi-paso; self-consistency muestrea
  varias cadenas y vota. La cadena no garantiza el cómputo interno real.
- **Prompt de sistema**: capas de rol, reglas, datos delimitados (defensa ante
  inyección), ejemplos y formato.
- **Salida estructurada**: escalera pedir JSON → prefill → validar esquema +
  reintentar → decodificación restringida (sintaxis garantizada). La semántica
  (datos alucinados) siempre requiere validación aparte.
