## 📖 Resumen de la materia

- **Subpalabras**: los LLM no ven palabras ni caracteres, sino unidades subpalabra
  de un vocabulario fijo (30k–250k). Frecuente → token único; raro → se descompone.
  Con base en bytes, no existe OOV.
- **BPE**: entrena fusionando iterativamente el par de símbolos adyacentes más
  frecuente y registra las *merges*; al tokenizar texto nuevo aplica las merges en
  el mismo orden.
- **Unigram/SentencePiece**: parte de un vocabulario grande, asume
  P(x) = Π P(tᵢ) y poda los tokens que menos verosimilitud aportan; segmenta con
  Viterbi y permite muestrear segmentaciones (regularización).
- **Consecuencias**: fertilidad (tokens/palabra) desigual entre idiomas, costo por
  token, contexto efectivo, y fallos de aritmética/deletreo se explican aquí.

Ejemplo clave: con merges aprendidas de {low, lower, newest, widest}, la palabra
nueva "lowest" se tokeniza `low` + `est_` sin haber sido vista nunca.
