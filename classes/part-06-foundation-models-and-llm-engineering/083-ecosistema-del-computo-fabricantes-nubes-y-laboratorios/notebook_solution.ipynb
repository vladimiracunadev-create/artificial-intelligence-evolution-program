## Solución de referencia

**Ejercicio 1.** Reformulación: *"hasta 3× el throughput agregado de una H100 en
un modelo propio de 7B, en FP8 con dispersión 2:4, a lote 512 y contexto de 256
tokens, según medición interna del fabricante"*. Con las cinco cualificaciones
puestas, la cifra sigue pudiendo ser cierta y deja de ser utilizable para decidir:
lo que falta es una corrida comparable con tu modelo y tu perfil de carga.

**Ejercicio 2.** Capa 1 fabricación: indirecta, sin alternativa a corto plazo.
Capa 2 acelerador: NVIDIA — proveedor único de facto por los kernels de vLLM.
Capa 3 infraestructura: un hiperescalar y una región — proveedor único y punto
único de fallo. Capa 4 modelo: pesos abiertos — **no** hay proveedor único, es la
capa sana. Resultado: **3 de 4** con proveedor único. La más barata de reducir es
la capa 3 (segunda región o segunda nube: coste de egreso y de operación); la
capa 2 exigiría validar ROCm o una capa de compilación y rehacer las evals.

**Ejercicio 3.** (a) `current` — hay fuente primaria verificable, aunque sea de
parte interesada. (b) `current` — reglas públicas y auditoría de terceros; es el
mejor tipo de evidencia disponible. (c) `unverified` — sin método ni datos
reproducibles. (d) `needs_replication` — hay método publicado pero nadie lo ha
reproducido. El criterio no es la simpatía por la fuente sino si la afirmación
viene acompañada de evidencia inspeccionable.

**Ejercicio 4.** El laboratorio deja fuera la afirmación cuya `evidence` es
`null`: la madurez se decide por la existencia de evidencia, no por lo llamativo
del enunciado. Es el mismo filtro que esta clase pide aplicar a los anuncios de
hardware.
