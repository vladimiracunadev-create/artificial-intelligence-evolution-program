## 📖 Resumen de la materia

- **Herramienta** = nombre + descripción (¡es prompt!) + JSON Schema de parámetros.
  El modelo la lee para decidir cuándo llamarla y con qué argumentos.
- **Bucle**: el modelo *propone* {name, input}; el runtime valida esquema y
  permisos, ejecuta, y devuelve el resultado como mensaje; el modelo continúa o
  responde. Límite de pasos y timeouts obligatorios.
- **Ejecución controlada**: mínimo privilegio, lectura vs escritura (confirmación
  humana para lo irreversible), validación de negocio, sandbox y auditoría. Una
  llamada propuesta es input NO confiable — la inyección de prompt es el ataque
  característico.
- **MCP** estandariza el descubrimiento y consumo de herramientas entre clientes y
  servidores (JSON-RPC), evitando N×M integraciones.
