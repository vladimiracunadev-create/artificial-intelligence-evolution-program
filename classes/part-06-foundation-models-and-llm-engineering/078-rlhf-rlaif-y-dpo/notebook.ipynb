## 📖 Resumen de la materia

- **RLHF (InstructGPT)**: SFT → modelo de recompensa entrenado con pares ordenados
  → PPO maximizando r con penalización β·KL contra la referencia.
- **Bradley–Terry**: P(y_w ≻ y_l) = σ(r_w − r_l); solo importan diferencias de
  recompensa, y comparar es más fiable que puntuar.
- **KL penalty**: ancla la política al modelo SFT; sin ella, reward hacking y
  degeneración. β regula el compromiso explorar/conservar.
- **DPO**: usando la forma cerrada del óptimo, la recompensa se vuelve implícita
  (r̂ = β·log π_θ/π_ref) y el alineamiento se reduce a una pérdida supervisada
  sobre pares — 2 modelos en memoria en vez de 4, sin PPO.
- **RLAIF**: misma matemática con un LLM juez como fuente de preferencias.
