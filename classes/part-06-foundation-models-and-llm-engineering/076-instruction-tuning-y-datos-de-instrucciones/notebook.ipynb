## 📖 Resumen de la materia

- **SFT**: continúa el next-token sobre pares (instrucción → respuesta) con
  *loss masking* (la pérdida solo cuenta sobre los tokens de la respuesta) y una
  plantilla de chat idéntica en entrenamiento e inferencia.
- **FLAN**: entrenar con 1 800+ tareas con plantillas variadas produce
  generalización zero-shot a tareas nunca vistas — el modelo aprende el meta-patrón
  "leer instrucción y ejecutarla".
- **Self-Instruct**: el propio LLM amplía ~175 semillas humanas generando
  instrucciones y respuestas, con filtrado por similitud y validez; barato pero
  hereda los errores del generador.
- **Calidad > cantidad**: LIMA logra calidad competitiva con 1 000 ejemplos
  curados; SFT enseña formato y comportamiento, no conocimiento nuevo.
