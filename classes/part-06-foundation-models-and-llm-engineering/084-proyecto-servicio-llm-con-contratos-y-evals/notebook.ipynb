## 📖 Resumen de la materia

- **Contrato**: el LLM se encapsula tras entrada/salida tipadas con validación en
  ambas direcciones y política de fallo (reintento → fallback → degradación
  honesta con `requiere_humano`); permite cambiar modelo/prompt sin romper
  consumidores.
- **Evals**: jerarquía programática → exact match → LLM-judge auditado → humano;
  golden set versionado con casos frontera; ningún cambio se despliega sin eval de
  regresión contra línea base, mirando métricas POR CLASE ponderadas por costo de
  error.
- **Observabilidad**: por request se registra versión de prompt/modelo, tokens,
  costo, TTFT/TPOT, validez y escalamiento; alertas por tasa de JSON válido,
  deriva de distribución y costo.
- **Principio**: la respuesta honesta declara confianza y limitaciones — igual que
