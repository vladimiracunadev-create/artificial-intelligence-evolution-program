## 📖 Materia resumida

**Sistema de IA evolutivo** = mejorable + verificable + reversible. Mejorable: hay un
canal por el que la experiencia produce cambios. Verificable: cada cambio pasa una
suite de evaluación antes de llegar a usuarios (sin esto, "evolutivo" = deriva).
Reversible: todo cambio es un artefacto versionado con rollback.

Arquitectura que integra las 15 partes:

```text
usuario → guardias de entrada (13) → orquestador/agente (9-10)
          ├ recuperación y memoria (8)   ├ modelos (4-7)
          ├ núcleo determinista (1-3)    └ herramientas (11-12)
        → guardias de salida (13) → respuesta + evidencia citada
        → telemetría → evaluación → cambio versionado (12-13)
```

Regla de composición: los componentes aprendidos **proponen**, el núcleo determinista
y el humano **disponen**. Cuatro trade-offs a declarar: autonomía↔control,
capacidad↔costo, recuerdo↔privacidad, novedad↔estabilidad.


## 🧮 Mecanismo clave: los 7 puntos del capstone

El entregable central no es el código, es el documento de decisiones:

```text
1. Caso acotado + desenlace medible + baseline NO-IA
2. Composición: qué parte es aprendida y qué parte decide de forma determinista
3. Guardias de entrada y salida (datos ≠ instrucciones; nada sin cita)
4. Suite de evaluación: casos de éxito, fuera-de-corpus, adversariales
5. Contrato de salida: {kind, evidence, limitations, seed, version}
6. Trade-offs declarados (autonomía, costo, privacidad, estabilidad)
7. Canal de evolución: cómo un fallo entra a la suite y cómo se promueve un cambio
```

Los casos "fuera de corpus" son los más informativos: miden si el sistema sabe decir
"no encontrado" en vez de inventar.
