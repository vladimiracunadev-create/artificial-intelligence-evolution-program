## 📖 Materia resumida

**ε-privacidad diferencial** (Dwork 2006): un mecanismo `M` es ε-DP si para datasets
vecinos `D, D'` (difieren en una persona) y toda salida `S`:
`Pr[M(D) ∈ S] ≤ e^ε · Pr[M(D') ∈ S]`. La garantía es del mecanismo y resiste
cualquier dato auxiliar — a diferencia de la anonimización clásica.

**Mecanismo de Laplace**: publicar `f(D) + Lap(Δf/ε)` donde `Δf` es la sensibilidad
(máximo cambio de `f` por un individuo). Conteo → `Δf = 1`. El presupuesto se
**compone**: k consultas de ε gastan k·ε.

**FedAvg** (arXiv:1602.05629): rondas de (enviar modelo global → E épocas locales en
cada cliente → media ponderada por n_k). Los datos no viajan, pero los **gradientes
filtran** (deep leakage): la privacidad formal exige SecAgg + DP encima.


## 🧮 Mecanismo clave: calibrar el ruido a mano

Conteo real `f(D) = 42`, sensibilidad `Δf = 1`, presupuesto `ε = 0.5`:

```text
b = Δf/ε = 2          → publicar 42 + Lap(2)
error típico E|Lap(2)| = 2   (≈ 5 % sobre 42)
cociente de densidades ≤ e^0.5 ≈ 1.65  → garantía cumplida
3 consultas de ε=0.5  → gasto total ε = 1.5
```

Menos ruido ⇒ más ε ⇒ menos privacidad. La calibración es exacta, no heurística.
