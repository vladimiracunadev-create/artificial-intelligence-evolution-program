## 📖 Materia resumida

Tres casos que separan promesa de resultado:

- **AlphaFold 2** (DOI 10.1038/s41586-021-03819-2): explota co-evolución en
  alineamientos múltiples + sesgo inductivo geométrico; validado **ciego** en CASP14
  (GDT_TS ~92, precisión experimental) y con incertidumbre por residuo (pLDDT).
- **GraphCast** (Science 2023): GNN entrenada en 40 años de ERA5, supera a HRES del
  ECMWF a 10 días en ~1 minuto. Pero depende de ERA5, los extremos son cola escasa
  y es meteorología, **no proyección climática**.
- **Salud**: Wynants et al. (BMJ 2020) revisaron 200+ modelos COVID y no recomendaron
  ninguno para uso clínico. Causas: fuga/atajos, sin validación externa, métrica
  equivocada, sin efecto en desenlaces, sesgo por subgrupo.

Lo que distingue los casos no es la arquitectura sino el **régimen de validación**:
ciega, externa, prospectiva, con incertidumbre reportada.


## 🧮 Mecanismo clave: AUC alto, VPP bajo

Cribado con prevalencia 1 %, sensibilidad 95 %, especificidad 90 %, 10,000 personas:

```text
TP = 95     FP = 990     FN = 5     TN = 8,910
VPP = 95/1,085 ≈ 8.8 %      VPN ≈ 99.94 %
```

De 100 positivos, ~9 tienen la condición. El modelo sirve para **descartar** (VPN
altísimo), no para diagnosticar. La utilidad depende de prevalencia, costos
asimétricos y existencia de prueba confirmatoria: nada de eso aparece en el AUC.
