## 📖 Materia resumida

Tres generaciones de IA para código, con contratos de evaluación distintos:

1. **Autocompletado** (Codex, arXiv:2107.03374): función aislada + tests ocultos
   (HumanEval), métrica **pass@k**.
2. **Chat**: explicación/refactor con el humano como integrador y juez.
3. **Agente de repo** (SWE-bench, arXiv:2310.06770): issue real de GitHub + repo
   completo; se acepta si los tests del fix real pasan (FAIL_TO_PASS) sin romper los
   existentes (PASS_TO_PASS).

El bucle del agente: localizar → editar → ejecutar tests → leer traceback → iterar.
La señal de los tests es lo que convierte generación probabilística en ingeniería.

**Modernización de legado**: primero tests de caracterización (fijan el
comportamiento actual, bugs incluidos), luego traducción por unidades verificada
contra ellos. Legacy code = código sin tests (Feathers).


## 🧮 Mecanismo clave: pass@k sin sesgo

Con `n` muestras y `c` correctas: `pass@k = 1 − C(n−c,k)/C(n,k)`.

```text
n=5, c=2:
pass@1 = 1 − 3/5   = 0.40
pass@3 = 1 − 1/10  = 0.90     (ingenuo 1−0.6³ = 0.784: sesgado)
```

Muestrear k veces y filtrar por tests convierte 40 % por intento en 90 % — solo si
existen tests que filtren. Sin verificador, más muestras solo dan más candidatos
plausibles.
