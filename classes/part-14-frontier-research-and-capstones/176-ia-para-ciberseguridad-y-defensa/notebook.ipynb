## 📖 Materia resumida

La IA defensiva cubre cuatro funciones: **detección** (firmas: precisas, ciegas a lo
nuevo; anomalías: cubren lo nuevo, inundan de falsos positivos), **triaje** (priorizar
alertas — el uso más maduro de LLM: resumir, correlacionar con MITRE ATT&CK),
**respuesta** (contener; peligrosa de automatizar) e **inteligencia**.

Resultado central — **falacia de la tasa base** (Axelsson):

```text
P(intrusión | alerta) = TPR·p / (TPR·p + FPR·(1−p))
```

Con prevalencia p minúscula, el FPR domina: reducir falsas alarmas vale más que
aumentar detección. Y el dominio es **adversarial**: el atacante prueba contra la
defensa (evasión), el tráfico cambia (drift), y el contenido del atacante que se
pasa a un LLM es un vector de prompt injection.


## 🧮 Mecanismo clave: la tasa base en números

1,000,000 eventos/día, 100 maliciosos (p = 10⁻⁴), TPR = 0.99, FPR = 0.01:

```text
TP = 99        FP = 9,999
P(malicioso | alerta) = 99/10,098 ≈ 1 %
con FPR = 0.001:  99/1,099 ≈ 9 %   (mejora ×9)
con TPR = 0.999:  ≈ 1 %            (mejora ~nada)
```

El "detector del 99 %" produce 10,098 alertas con 99 % de falsas alarmas. La métrica
operativa es la precisión de la cola que el analista puede revisar.
