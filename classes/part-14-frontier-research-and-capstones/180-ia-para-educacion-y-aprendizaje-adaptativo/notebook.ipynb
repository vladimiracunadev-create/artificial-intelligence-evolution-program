## 📖 Materia resumida

**Problema 2 sigma** (Bloom 1984): la tutoría 1-a-1 mejora ~2σ sobre la clase
tradicional; el reto es lograrlo a costo escalable. Un tutor adaptativo resuelve
tres problemas: modelar al estudiante, elegir la siguiente actividad (en la **ZPD**:
éxito esperado 0.6-0.8) y andamiar sin dar la solución.

**Bayesian Knowledge Tracing** (Corbett & Anderson 1995): habilidad binaria latente
con 4 parámetros — P(L0) inicial, P(T) aprender, P(S) slip, P(G) guess:

```text
P(L|correcto) = P(L)(1−S) / [P(L)(1−S) + (1−P(L))G]
P(L')         = P(L|obs) + (1−P(L|obs))·P(T)
```

**DKT** (RNN, arXiv:1506.05908) captura correlaciones entre habilidades pero pierde
interpretabilidad. Los LLM aportan diálogo y explicación; fallan por complacencia,
alucinación pedagógica y ausencia de modelo de estudiante persistente. La métrica
válida es **learning gain** (pre/post test), no satisfacción.


## 🧮 Mecanismo clave: una actualización BKT

P(L)=0.40, S=0.10, G=0.20, T=0.15; respuesta correcta:

```text
evidencia:  0.40·0.90 / (0.36 + 0.60·0.20) = 0.36/0.48 = 0.75
transición: 0.75 + 0.25·0.15 = 0.7875
```

el tutor sigue dando práctica.
