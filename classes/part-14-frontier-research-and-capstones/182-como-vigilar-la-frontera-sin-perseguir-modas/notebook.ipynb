## 📖 Materia resumida

Separa lo que aprendes en dos capas con vidas medias distintas: **núcleo estable**
(probabilidad, optimización, búsqueda, evaluación, ingeniería — décadas) y
**frontera** (APIs, nombres de modelos, frameworks, benchmarks — meses). El núcleo se
estudia; la frontera se **vigila** con un proceso barato.

**Niveles de madurez** por señales observables: `emerging` (1-2 papers, sin
evaluación externa → anotar), `current` (implementaciones múltiples y uso externo →
prototipo acotado), `standard` (spec pública, adopción multi-proveedor → adoptar si
resuelve un problema real), `deprecated` (migrar y documentar).

**Cinco preguntas de filtrado**: ¿contra qué baseline? ¿quién evaluó? ¿hay
contaminación del benchmark? ¿es reproducible? ¿cuánto cuesta?

Este repositorio implementa el patrón en `frontier/current-topics.yaml`: cada tema
declara `maturity`, `reviewed` (fecha) y `source` (enlace primario).


## 🧮 Mecanismo clave: rúbrica de 4 criterios

2 puntos por criterio cumplido, 1 parcial, 0 ausente — baseline, evaluador
independiente, reproducibilidad, costo declarado:

```text
A. Spec pública + 3 implementaciones      8/8 → standard: adoptar si aplica
B. +2 % en benchmark público, autoeval    2/8 → emerging: anotar y revisar en 8 semanas
C. Framework usado por 2 empresas ajenas  4/8 → current: prototipo con criterio de salida
D. +2 % con 1000× cómputo, código cerrado 5/8 → hallazgo sobre escalado, no sobre método
```

El filtro es barato por diseño: ninguna de las cuatro decisiones exige leer el paper
completo.
