## 📖 Materia resumida

**Escalera de Pearl**: (1) asociación P(Y|X) — lo que el ML estándar estima;
(2) intervención P(Y|do(X)) — el efecto de actuar; (3) contrafactuales — qué
habría pasado. Subir de peldaño exige un **modelo causal** (DAG con supuestos) o
experimentos; ningún volumen de datos observacionales basta por sí solo.

**Puerta trasera**: si Z bloquea los caminos traseros entre X e Y,
`P(Y|do(X)) = Σ_z P(Y|X,Z=z)·P(Z=z)`. El **do-cálculo** (3 reglas, completo)
decide en general qué efectos son identificables.

**Descubrimiento causal**: PC/FCI (independencias), GES (score), LiNGAM
(asimetrías funcionales). Sin supuestos fuertes solo se identifica la clase de
equivalencia de Markov: X→Y y X←Y suelen ser indistinguibles sin intervenir.
