## 📖 Materia resumida

**IA neuro-simbólica**: combinar redes neuronales (percepción aprendida, tolerante
al ruido, sin garantías) con razonamiento simbólico (deducción exacta,
verificable, frágil ante entrada cruda). La división de trabajo: **la red percibe,
el símbolo deduce**, con una interfaz probabilística entre ambos.

**Taxonomía de Kautz** (AAAI 2020 / AI Magazine 2022), seis patrones:

| Tipo | Patrón | Ejemplo |
|---|---|---|
| 1 | Symbolic Neuro symbolic | LLM: texto → red → texto |
| 2 | Symbolic[Neuro] | AlphaGo: MCTS llama a una red evaluadora |
| 3 | Neuro \| Symbolic | Concept Learner: percepción + programa |
| 4 | Symbolic → Neuro | reglas compiladas como pérdida de entrenamiento |
| 5 | Neuro_{Symbolic} | lógica tensorizada dentro de la red |
| 6 | Neuro[Symbolic] | LLM que invoca solver / calculadora / verificador |


## 🧮 Mecanismo clave (estilo DeepProbLog)

1. La red produce `P(símbolo | entrada)` — percepción incierta.
2. Una regla lógica dura define qué mundos son válidos.
3. `P(conclusión) = suma de P(mundos que la satisfacen)`, renormalizada.

Ejemplo a mano: A: P(3)=0.7, P(5)=0.3; B: P(2)=0.6, P(7)=0.4; regla "suma=10".
El mundo más probable (3,2) viola la regla; el único válido es (3,7) con
p = 0.28, que tras renormalizar queda en 1.0. La regla corrige la percepción.

El laboratorio de abajo usa la parte simbólica del patrón: una regla dura e
inspeccionable que separa afirmaciones con evidencia de afirmaciones sin ella.
