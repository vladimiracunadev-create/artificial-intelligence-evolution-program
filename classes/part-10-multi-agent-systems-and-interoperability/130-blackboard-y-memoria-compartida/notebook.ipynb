## 📖 Materia resumida

**Blackboard** (Nii, 1986; Hearsay-II): memoria compartida jerárquica de *hipótesis*
(autor, nivel, confianza, versión) + **fuentes de conocimiento** (KS) con condición de
activación y acción, que no se conocen entre sí + **control** que elige qué KS ejecuta
en cada ciclo.

**Flujo oportunista**: no hay plan a priori; la secuencia emerge de qué hipótesis
aparecen (coordinación indirecta / estigmergia). Contraste con supervisor-workers:
allí la descomposición se decide de antemano.

**Conflictos**: hipótesis incompatibles *coexisten* con confianza (no se pisan);
lecturas obsoletas exigen versionado; ciclos sin progreso = deadlock epistémico →
escalar. En LLM se añade el límite de ventana: vistas resumidas, no volcado completo.


## 🧮 Ejemplo de referencia

Incidente con 3 KS (logs, metrics, causal) y niveles síntoma → hipótesis → causa:

```text
t1 logs:    "502 desde 09:31" (0.9)      t2 metrics: "p99 ×8" (0.85)
t3 causal:  "saturación del pool" (0.6)  ← se activó al ver 2 síntomas correlacionados
t4 logs:    "pool exhausted en BD" (0.95) ← búsqueda DIRIGIDA por la hipótesis de otra KS
t5 causal:  confianza 0.9 ≥ umbral → fin
```

t4 es la esencia del patrón: una KS orientó a otra sin mensaje directo alguno.
