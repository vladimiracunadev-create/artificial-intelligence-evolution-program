## 📖 Materia resumida

**Agent Skill**: carpeta portable con `SKILL.md` (frontmatter `name` + `description`
+ cuerpo de instrucciones) más scripts y recursos opcionales. La description es la
interfaz pública: declara qué hace y cuándo activarlo.

**Divulgación progresiva**: nivel 1 metadatos (siempre en contexto, ~decenas de
tokens), nivel 2 cuerpo (si la tarea coincide), nivel 3 archivos/scripts (solo si
hacen falta). Cientos de skills instalados ≠ cientos de skills pagados.

**Instrucciones vs. scripts**: lo que debe ser siempre igual → script (determinista,
barato); lo que exige criterio → instrucciones. **Seguridad**: instalar un skill de
terceros = inyectar prompts y código; se audita como software, no como texto.
