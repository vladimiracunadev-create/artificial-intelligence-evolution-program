## 📖 Materia resumida

**A2A** (Google 2025, hoy en la Linux Foundation): protocolo para que agentes de
distintas organizaciones colaboren como **cajas opacas** — sin compartir modelo,
memoria ni herramientas.

**Descubrimiento**: la **Agent Card** (`/.well-known/agent-card.json`) declara
identidad, endpoint, skills con input/outputModes y auth — el contrato de la clase
128 hecho protocolo, legible *antes* del primer mensaje.

**Trabajo**: **tasks** con ciclo `submitted → working → [input-required] →
completed/failed/canceled` sobre JSON-RPC/HTTP+SSE; la conversación viaja en
**messages** y el entregable en **artifacts** (resultado ≠ charla). Mnemónico
oficial: MCP conecta al agente con sus *herramientas*; A2A con *otros agentes*.


## 🧮 Ejemplo de referencia

Delegación de verificación de antecedentes a un agente externo:

```text
1 descubrir: GET agent-card → skill "background-check", auth bearer
2 delegar:   message/send → Task T-77 (working)
3 interactuar: ← input-required ("¿alcance internacional?") → aclaración
4 resultado: ← completed + Artifact {result: "sin_hallazgos", fuentes: 4}
```

No viajó ni el modelo, ni las fuentes, ni el prompt del remoto: solo tarea, mensajes
y artefacto tipado.
