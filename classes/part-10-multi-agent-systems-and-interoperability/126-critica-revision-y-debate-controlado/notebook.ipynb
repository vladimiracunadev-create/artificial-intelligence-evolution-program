## 📖 Materia resumida

**Generador-crítico**: roles con incentivos opuestos; el crítico necesita rúbrica,
contexto limpio y, idealmente, modelo/prompt distinto (auto-revisión hereda los mismos
puntos ciegos). Bucle acotado por `max_rounds`.

**Debate (Du et al., arXiv:2305.14325)**: n agentes responden independientemente;
r rondas leyendo las respuestas ajenas y revisando la propia; final por consenso o
voto. Mejora aritmética y factualidad; coste n·r; converger ≠ acertar.

**Votación (Condorcet)**: con votantes *independientes* de p > 0.5, la mayoría acierta
más que el individuo. El voto reduce **varianza**, no **sesgo**: réplicas del mismo
modelo están correlacionadas y la ganancia real es menor que la teórica.


## 🧮 Ejemplo de referencia

3 votantes independientes con p = 0.7:

```text
P(mayoría) = 0.7³ + 3·0.7²·0.3 = 0.343 + 0.441 = 0.784
n = 5 → 0.837  (+5.3 pts por 2 llamadas más; retorno decreciente)
correlación total → P = 0.7  (el voto no compró nada)
```

Por eso el debate añade *rondas de argumentos* al voto: los argumentos cruzados pueden
corregir el error sistemático; el voto puro, no.
