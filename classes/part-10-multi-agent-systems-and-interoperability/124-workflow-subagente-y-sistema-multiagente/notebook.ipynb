## 📖 Materia resumida

**Workflow**: el *código* fija el flujo; el LLM rellena pasos (Anthropic, "Building
effective agents"). **Subagente**: agente invocado por otro agente como herramienta,
con contexto aislado y contrato de retorno. **Sistema multiagente**: n agentes
autónomos que se coordinan por mensajes o memoria compartida (Wooldridge, 2009).

**Delegación** = subtarea + presupuesto + contrato de retorno. La pregunta clave no es
"¿cuántos agentes?" sino "¿quién decide el siguiente paso y cuántos contextos existen?".

**Cuándo NO multiagente**: descomposición fija → workflow; solo aislar contexto →
subagente; sin paralelismo real → agente único. Anthropic mide ≈15× tokens frente a un
chat simple: la arquitectura debe pagarse con valor.
