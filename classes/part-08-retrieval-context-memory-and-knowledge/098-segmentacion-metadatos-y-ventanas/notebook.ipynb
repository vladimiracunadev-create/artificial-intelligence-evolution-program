## 📖 Materia resumida

**Chunking**: partir documentos en fragmentos indexables. Tensión central: chunks
pequeños → embeddings nítidos pero sin contexto; grandes → contexto pero vectores
"promedio" que recuperan peor.

**Estrategias**: tamaño fijo (+ solape `S`), recursiva (secciones→párrafos→oraciones),
estructural (fronteras del formato), semántica (corte al caer la similitud entre
oraciones). Con tamaño fijo, cada chunk avanza `N − S` tokens.

**Ventanas**: sentence-window y parent-document separan la *unidad de búsqueda*
(pequeña, precisa) de la *unidad de lectura* (grande, con contexto).

**Metadatos** (`doc_id`, sección, fecha, permisos): habilitan filtrar, atribuir y
depurar — tres cosas que el vector no puede hacer solo.


## 🧮 Ejemplo de referencia

Documento de 900 tokens, `N = 300`, `S = 50` → paso efectivo 250:

```text
chunk 0: [0, 300)   chunk 1: [250, 550)   chunk 2: [500, 800)   chunk 3: [750, 900)
⌈(900−50)/250⌉ = 4 chunks · 1050 tokens almacenados → sobrecoste 1.17×
```

El solape reduce (no elimina) el riesgo de cortar una idea por la mitad; solo las
fronteras estructurales lo evitan por construcción.
