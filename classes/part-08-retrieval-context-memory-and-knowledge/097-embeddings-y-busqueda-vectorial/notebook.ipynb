## 📖 Materia resumida

**Embedding**: función `E: texto → ℝ^d` aprendida por contraste; la cercanía geométrica
codifica similitud semántica.

**Similitud coseno**: `cos(u,v) = (u·v)/(‖u‖‖v‖)`; rango [−1, 1]; invariante a escala.
Con vectores normalizados, coseno ≡ producto punto y `‖u−v‖² = 2 − 2·cos(u,v)`.

**Búsqueda exacta** (flat): `O(n·d)` por consulta, recall 100 %.
**ANN**: IVF (celdas k-means + `nprobe`), PQ (compresión por subvectores),
**HNSW** (grafo jerárquico de mundos pequeños, arXiv:1603.09320): descenso greedy por
capas, `O(log n)` empírico, recall controlado por `efSearch`. La métrica de un índice
ANN es siempre el par **recall@k / latencia**, nunca uno solo.


## 🧮 Ejemplo de referencia

Con `q=(1,2,2)`, `d1=(2,4,4)`, `d2=(0,3,4)`, `d3=(2,−1,2)`:

```text
cos(q,d1) = 18/(3·6) = 1.000    (d1 = 2q: misma dirección, distinta magnitud)
cos(q,d2) = 14/(3·5) ≈ 0.933
cos(q,d3) =  4/(3·3) ≈ 0.444    →  ranking d1 > d2 > d3
```

Verifica los tres cosenos a mano antes de ejecutar el laboratorio: son las mismas
operaciones que un índice vectorial repite millones de veces.
