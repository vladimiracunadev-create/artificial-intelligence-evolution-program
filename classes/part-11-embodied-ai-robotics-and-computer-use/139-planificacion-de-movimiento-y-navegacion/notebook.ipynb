## 📖 Resumen de la materia

El **espacio de configuración** reduce el robot a un punto inflando los
obstáculos con su geometría; planificar es buscar un camino en `C_libre`.

- **A\***: expande por `f = g + h`; con heurística admisible (nunca
  sobreestima, p. ej. Manhattan en rejilla de 4 vecinos) es completo y óptimo.
  `h=0` ⇒ Dijkstra; `h` inflada ⇒ rápido pero subóptimo. No escala en
  dimensión (maldición de la dimensionalidad).
- **RRT/RRT\***: muestrean el C-space y hacen crecer un árbol;
  probabilísticamente completos; RRT no es óptimo, RRT* converge al óptimo.
  Son la opción en 6D+ (brazos).
- **DWA**: capa local a ~20 Hz; considera solo velocidades alcanzables
  (ventana dinámica), simula trayectorias cortas de `(v, ω)`, descarta las que
  colisionan y puntúa avance/holgura/velocidad.

El patrón operativo (Nav2) es global (A*, ~1 Hz) + local (DWA, ~20 Hz): la
misma jerarquía deliberativo/reactivo de la clase 136.
