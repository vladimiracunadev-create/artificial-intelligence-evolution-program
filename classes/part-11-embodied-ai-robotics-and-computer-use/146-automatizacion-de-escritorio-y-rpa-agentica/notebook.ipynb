## 📖 Resumen de la materia

La **RPA clásica** automatiza UI con flujos deterministas + **selectores**
(expresiones que identifican controles). Es imbatible en procesos estables:
barata, determinista, auditable. Su talón de Aquiles: el selector describe la
UI del día de la grabación — cualquier cambio la rompe (mantenimiento
creciente) y ante ambigüedad solo sabe lanzar excepción.

La **RPA agéntica** inserta modelos en nodos concretos: percepción robusta
(localizar elementos por semántica cuando el selector muere), decisión bajo
ambigüedad (clasificar/extraer formatos nuevos) y manejo de excepciones. El
precio: no determinismo, coste por inferencia y fallos nuevos (alucinación
con formato válido, inyección en documentos).

Regla de diseño de la clase: **columna vertebral determinista, agente solo en
la variabilidad irreducible**, con validaciones duras, umbral de confianza,
escalado a humano y muestreo continuo. Ejemplo: 500 facturas/mes, 16 % no
estándar → el nodo agéntico convierte 8 h/mes de excepciones en ~48 min +
revisión, sin tocar los 420 casos que ya eran deterministas.
