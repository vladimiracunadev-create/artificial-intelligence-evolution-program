## 📖 Resumen de la materia

El controlador convierte el error `e = r − y` en acción:
`u = Kp·e + Ki·∫e + Kd·de/dt`. Cada término tiene un papel causal: **P**
agresividad (solo-P deja error estacionario ante perturbación constante),
**I** elimina el error final (riesgo: windup y sobreimpulso), **D** amortigua
(riesgo: amplificar ruido).

Ejemplo de la clase (`v' = u − 0.5v`, referencia 10): con `Kp=1` el equilibrio
queda en 6.67 (error 3.33); con `Kp=5` en 9.09 (el error se reduce pero no
desaparece); con PI (`Kp=2, Ki=1`) la integral aporta el `u=5` que pide la
fricción y el error estacionario se anula; añadir `Kd=0.5` recorta el
sobreimpulso.

El **control aprendido** (políticas RL tipo PPO/SAC) maneja no linealidades y
contactos que el PID no puede, a cambio de perder garantías formales y
requerir simulación masiva + transferencia sim-to-real. El patrón comercial es
híbrido: la política decide consignas, PIDs articulares ejecutan, y una
envolvente de seguridad clásica vigila.
