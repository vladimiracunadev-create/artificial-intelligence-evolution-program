## 📖 Resumen de la materia

El computer use por visión aplica el lazo de la clase 136 al escritorio:
**observar** (screenshot), **razonar**, **actuar** (`click(x,y)`, `type`,
`key`, `scroll`), **verificar** (nuevo screenshot contra una postcondición).
No hay propiocepción: todo se ve, nada se siente; y la UI es asíncrona, así
que se espera estabilidad en lugar de dormir tiempos fijos.

- **Grounding**: traducir "clic en Guardar" a coordenadas. Es el eslabón débil
  medible (ScreenSpot, OSWorld). Con acierto p por clic, una tarea de n clics
  se completa con probabilidad `p^n`: 0.95¹⁵ ≈ 46 % — la aritmética del
  horizonte de la clase 141.
- **Mapeo de resolución**: las coordenadas del modelo viven en el screenshot
  reescalado; olvidarse del factor (p. ej. ×1.5 de 1280×720 a 1920×1080)
  produce clics sistemáticamente desviados con grounding perfecto.
- **Seguridad**: el escritorio ejecuta efectos irreversibles y el texto en
  pantalla es entrada no confiable (inyección visual). Mínimos: sandbox,
  acciones con confirmación, y evidencia screenshot-acción-resultado.
