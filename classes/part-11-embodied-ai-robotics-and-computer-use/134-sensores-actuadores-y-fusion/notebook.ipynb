## 📖 Resumen de la materia

Los sensores tienen **sesgo** (se corrige con calibración) y **varianza** (se
gestiona con fusión); los actuadores tienen holgura, saturación y retardo. La
fusión óptima de dos mediciones pondera por la inversa de la varianza y
siempre produce menos incertidumbre que el mejor sensor solo.

El **filtro de Kalman 1D** mantiene una creencia `N(x̂, P)` y alterna:

```text
Predicción:  x̂⁻ = a·x̂ + b·u ;  P⁻ = a²·P + Q      (la incertidumbre crece)
Corrección:  K = P⁻/(P⁻+R) ;  x̂ = x̂⁻ + K·(z−x̂⁻) ;  P = (1−K)·P⁻
```

La ganancia K ∈ (0,1) reparte la confianza entre modelo y sensor: R pequeño ⇒
K→1 (manda el sensor); R grande ⇒ K→0 (manda el modelo). El **filtro
complementario** (`α·giro + (1−α)·acel`) es la alternativa sin covarianzas
para errores separables en frecuencia. Ejemplo de la clase: con `x̂=0, P=1,
Q=0.25, R=1, u=1, z=1.2` se obtiene `K=0.556, x̂=1.111, P=0.556`.
