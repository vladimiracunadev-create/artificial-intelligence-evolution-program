## 📖 Resumen de la materia

Un agente encarnado opera en lazo cerrado: **percepción** (sensores → estado
estimado, siempre ruidoso), **planificación** (estado + objetivo → decisión),
**acción** (decisión → actuadores) y **feedback** (el resultado vuelve por los
sensores). Tres arquitecturas organizan ese lazo:

- **Sense-Plan-Act** (Shakey): tubería secuencial con modelo central. Óptima
  sobre su modelo, pero lenta y frágil ante cambios.
- **Subsumption** (Brooks, 1986): capas reactivas sensor→actuador sin modelo
  central; las capas altas inhiben a las bajas. Robusta y rápida, sin
  razonamiento de largo plazo.
- **Híbrida de tres capas**: deliberativa (~1 Hz), ejecutiva (~10 Hz) y
  reactiva (~1000 Hz). La seguridad vive siempre en la capa rápida. Es el
  patrón de ROS 2 + Nav2.

En el ejemplo del aspirador 1D: el plan óptimo (6 movimientos) se invalida con
un obstáculo móvil; el diseño reactivo hace 8 movimientos pero nunca se
detiene; el híbrido logra ~6-7 con robustez.
