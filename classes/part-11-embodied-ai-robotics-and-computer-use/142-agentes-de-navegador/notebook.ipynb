## 📖 Resumen de la materia

El navegador da al agente lo que el escritorio no: **DOM** (árbol completo de
elementos) y **árbol de accesibilidad** (roles + nombres: la representación
compacta ideal para un LLM), más acciones de alto nivel (`click(elemento)`,
`fill`, `goto`) vía Playwright/CDP con espera automática de accionabilidad.

- **DOM vs visión**: el DOM es preciso y barato pero miente ante canvas,
  shadow DOM e iframes; la visión lo ve todo pero paga grounding a píxel y
  tokens de imagen. Los agentes competitivos son híbridos.
- **Fallos típicos**: referencias a nodos muertas tras un re-render, esperas
  fijas en páginas asíncronas, bucles de reintento sin cambio de estrategia.
- **WebArena** (arXiv:2307.13854): 812 tareas sobre sitios reales
  auto-hospedados con verificación programática del estado final. Humanos
  ~78 %; el mejor agente del paper original ~14 %. Lección: solo cuenta el
  éxito end-to-end verificado — los pasos "razonables" no predicen nada.

Ejemplo de la clase: descargar la factura del pedido #1043 cuesta ~6 acciones
con 2 groundings difíciles por visión (éxito ≈ 78 %) y 3 acciones por DOM
(≈ 97 %) — salvo que la tabla sea canvas, donde solo la visión sobrevive.
