## 📖 Resumen de la materia

La simulación da **escala, seguridad, ground truth y reproducibilidad**, pero
crea el **reality gap**: la política sobreajusta a la física y las imágenes
sintéticas (componentes dinámico y perceptivo del gap).

- **Domain randomization** (arXiv:1703.06907): variar parámetros del entorno
  (fricción, masas, luces, latencias) en cada episodio para que la realidad
  caiga dentro de la distribución entrenada. Precio: política más conservadora.
- **Identificación de sistema**: medir el hardware y calibrar el simulador;
  se combina randomizando el residuo de incertidumbre en torno a lo medido.
- **Digital twin**: modelo virtual de *un activo concreto* sincronizado con su
  telemetría; sin sincronización no es gemelo, es un simulador con marketing.

Ejemplo de la clase (empujar un bloque, `μ_real=0.42` desconocido): sim fijo a
μ=0.6 → 40 % de éxito real; randomizar μ∈(0.3, 0.8) → ~88 %; medir
(`μ̂=0.44±0.05`) y randomizar fino → ~93 %. Regla: mide lo que puedas,
randomiza lo que no, nunca confíes en un valor puntual.
