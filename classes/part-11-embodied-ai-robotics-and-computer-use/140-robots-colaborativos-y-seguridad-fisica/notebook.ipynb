## 📖 Resumen de la materia

Un **cobot** comparte espacio con personas: la seguridad pasa de la exclusión
(jaulas) al diseño. Marco normativo: **ISO 10218** (robot e integración) e
**ISO/TS 15066** (colaboración), con cuatro modos: parada monitorizada,
guiado manual, **velocidad y separación (SSM)** y **limitación de potencia y
fuerza (PFL)** — el único que permite contacto, bajo umbrales biomecánicos.

Núcleo cuantitativo de SSM, la distancia protectora:

```text
S = v_h·(t_r + t_s) + v_r·t_r + d_frenado + C
```

Ejemplo de la clase: con `v_h=1.6, v_r=1.0, t_r=0.1, t_s=0.3, C=0.2` sale
`S=1.09 m`; bajando el robot a 0.25 m/s, `S=0.64 m` — reducir velocidad
encoge la zona de exclusión y habilita colaborar cerca.

Principios transversales: capas de seguridad **independientes** del software
de aplicación (la política aprendida comanda, el limitador certificado
custodia), **fail-safe** (todo fallo termina en estado seguro) y evaluación de
riesgos de la **aplicación completa**, no del componente.
