## 📖 Resumen de la materia

**Behavioral cloning**: tratar demostraciones `(s, a)` como aprendizaje
supervisado. Funciona… hasta que la política comete su primer error y entra en
estados que el experto nunca visitó — **covariate shift** — donde vuelve a
errar y se aleja más. El coste crece **O(ε·T²)** con el horizonte (Ross &
Bagnell); es un problema de distribución de datos, no de capacidad del modelo.

**DAgger** (arXiv:1011.0686) rompe el bucle: ejecuta la política actual,
pide al experto la acción correcta en los estados que *ella* visita (incluidos
sus errores), agrega al dataset y reentrena. Garantiza O(ε·T) a cambio de un
experto interactivo.

Ejemplo de la clase: en un corredor de 10 celdas con acierto 0.95 en
distribución y 0.5 fuera, BC completa `0.95²⁰ ≈ 36 %` de episodios de 20
pasos; una ronda de DAgger que etiqueta las celdas vecinas eleva la tasa por
encima del 85 % porque enseña a *recuperarse*, no solo a seguir la ruta ideal.
