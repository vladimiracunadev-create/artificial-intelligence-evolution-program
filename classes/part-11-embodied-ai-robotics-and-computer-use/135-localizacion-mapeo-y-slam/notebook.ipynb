## 📖 Resumen de la materia

**Localización** estima la pose con mapa conocido; **mapeo** construye el mapa
con pose conocida; **SLAM** estima ambos a la vez porque ninguno se conoce —
es un problema huevo-gallina resuelto con estimación conjunta
`p(pose, mapa | observaciones, controles)`.

- **Odometría**: integra el movimiento; su error se acumula sin cota. El error
  angular es el peor: desvía toda la trayectoria posterior (`desvío ≈ d·sin ε`).
- **EKF-SLAM**: estado aumentado `[pose, landmarks]` con covarianza completa;
  predicción-asociación-corrección; las correlaciones propagan cada corrección
  al mapa entero. Coste O(N²).
- **Cierre de lazo**: reconocer un lugar visitado añade una restricción que
  redistribuye la deriva acumulada; un cierre falso corrompe el mapa.

Ejemplo de la clase: recorriendo un cuadrado de 10 m con 1° de sesgo por giro,
el robot termina a ~1.2 m del origen; una observación de un landmark en el
origen reduce ese error al nivel del ruido del sensor.
