## Solución de referencia

**Ejercicio 1.**
a) **allowlist**: lectura pura, reversible, costo nulo.
b) **intermedia**: enviar es irreversible pero el borrador fue revisado; el
   costo depende del contenido → escalada con contexto.
c) **denylist**: destrucción definitiva de datos, irreversible por diseño.
d) **allowlist**: reversible trivialmente (renombrar de vuelta), costo bajo.
e) **denylist** (o intermedia con doble aprobación según política): movimiento
   de dinero a destinatario nuevo = irreversible y de alto costo.

**Ejercicio 2.**
(a) 18/20 = **90 %** de éxito. (b) 4/20 = **20 %** de episodios con propuesta
peligrosa. (c) 8/9 ≈ **89 %** de aprobación. La cifra que bloquea la
relajación es la **(b)**: 1 de cada 5 episodios el planificador propone algo
prohibido; la última línea de defensa está trabajando como primera. La (c),
tan alta, además sugiere revisar si el humano lee o aprueba por hábito.

**Ejercicio 3.**
