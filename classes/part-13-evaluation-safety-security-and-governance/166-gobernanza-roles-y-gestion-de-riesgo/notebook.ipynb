## 📖 Resumen de la materia

**Gobernanza** = roles, procesos y decisiones (con autoridad real para detener) que alinean un
sistema de IA con las políticas y la ley. El **NIST AI RMF** la organiza en cuatro funciones
continuas: **GOVERN** (transversal: roles, políticas, rendición de cuentas), **MAP** (contexto y
riesgos), **MEASURE** (cuantificar con evals/fairness/red team/calibración) y **MANAGE** (mitigar,
aceptar, transferir o evitar).

Las **tres líneas de defensa** separan con independencia creciente: 1ª = producto (dueño del
riesgo), 2ª = riesgo/legal/seguridad (supervisión), 3ª = auditoría (aseguramiento independiente). El
**riesgo residual** —el que queda tras mitigar— se **acepta explícitamente** con un responsable; un
riesgo ignorado no es gestión, es negligencia.


## 🧮 Mini-ejemplo: matriz de riesgo

Scoring de CV, P·I en escala 1-5:

```text
R1 sesgo de género   P=4 I=5 -> 20 crítico  -> mitigar (DI>=0.8 + revisión humana)
R3 fuga de PII        P=2 I=5 -> 10 alto     -> mitigar (redacción en logs)
R2 alucinación        P=3 I=2 ->  6 medio    -> aceptar y monitorear
```

Tras mitigar, R1 residual sigue *alto*: el risk owner decide despliegue **condicionado** con
aceptación firmada del residual. La gobernanza hace la decisión explícita, trazable y con dueño.
