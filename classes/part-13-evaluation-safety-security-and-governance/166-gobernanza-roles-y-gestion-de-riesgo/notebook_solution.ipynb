
Tratamiento sugerido: R1 → mitigar (grounding + abstención + derivación a médico, clase 165); R3 →
mitigar (minimización y cifrado, clase 162) y/o transferir parte vía seguro; R2 → mitigar

**Ejercicio 4.** (i) 1ª línea: el equipo de producto implementa el filtro. (ii) 2ª línea:
riesgo/legal/clínico define la política y el umbral de derivación. (iii) 3ª línea: auditoría interna
verifica que la derivación ocurre. (iii) debe ser independiente porque si lo verifica quien
construye u opera el sistema, hay conflicto de interés: el auditor tendería a confirmar su propio
trabajo, y el control perdería credibilidad ante el órgano de gobierno y los reguladores.


````python
# Verificación numérica del Ejercicio 3
riesgos = {"R1": (3, 5), "R2": (4, 2), "R3": (2, 5)}
niveles = {k: p * i for k, (p, i) in riesgos.items()}
orden = sorted(niveles, key=niveles.get, reverse=True)
print("niveles:", niveles)
print("prioridad:", orden)
assert niveles == {"R1": 15, "R2": 8, "R3": 10} and orden == ["R1", "R3", "R2"]
````
