## 📖 Resumen de la materia

**Sesgo** (fairness) es una diferencia sistemática de comportamiento entre grupos protegidos; entra
sobre todo por los **datos** (histórico, representación, medición), no por el algoritmo. Definiciones
de equidad de grupo: **paridad demográfica** (misma tasa de positivos), **igualdad de oportunidad**
(mismo TPR entre grupos), **equalized odds** (TPR y FPR iguales) y **calibración por grupo** (el
score significa lo mismo en cada grupo).

**Teorema de imposibilidad** (Kleinberg et al., arXiv:1609.05807): salvo tasas base iguales o
predicción perfecta, calibración e igualdad de FPR/FNR **no** pueden cumplirse a la vez — hay que
elegir un criterio y justificarlo. El **disparate impact** DI = P(Ŷ=1|desfavorecido)/P(Ŷ=1|favorecido);
la **regla del 80 %** marca indicio de impacto adverso si DI < 0.80.


## 🧮 Mini-ejemplo: disparate impact

Crédito, dos grupos de 500. Aprobados: A=200 (0.40), B=120 (0.24).

```text
DI = 0.24 / 0.40 = 0.60  ->  < 0.80  ->  indicio de impacto adverso contra B
```

Con la etiqueta real: TPR_A = 180/250 = 0.72 vs TPR_B = 96/150 = 0.64 → también falla igualdad de
oportunidad. Igualar un criterio moverá otro (imposibilidad de Kleinberg): la elección es normativa.
