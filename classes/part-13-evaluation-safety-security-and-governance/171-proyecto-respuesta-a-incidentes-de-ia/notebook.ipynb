## 📖 Resumen de la materia

Un **incidente de IA** es un evento en el que el sistema causa (o casi causa) un daño real:
exfiltración por inyección, decisión discriminatoria a escala, consejo peligroso, fuga de PII. El
**ciclo de respuesta** (adaptado de NIST SP 800-61): **preparación** (plan, roles, logs YA listos),
**detección y análisis** (confirmar, clasificar severidad, delimitar alcance), **contención** (parar
el daño con acciones *reversibles* sin destruir evidencia), **erradicación** (causa raíz),
**recuperación** (validar con regresión antes de re-abrir) y **post-incidente** (causa raíz +
convertir el caso en regresión permanente).

Tensión central: contener rápido vs **preservar la evidencia**. Preparación y post-incidente son las
fases más descuidadas y las que más reducen el daño futuro. Cada incidente cierra el ciclo con las
clases previas: detección (163-165), causa raíz (160-163), evidencia (170), prevención (158, 166).


## 🧮 Mini-ejemplo: exfiltración por inyección

Asistente de correo reenvía datos a un dominio externo:

```text
Detección   -> alerta de envío externo; inyección indirecta confirmada -> SEV1
Contención  -> desactivar send_email por flag + snapshot de logs + revocar credencial (reversible)
Erradicación-> allowlist de destinatarios + marcar contenido no confiable + aprobación humana
Recuperación-> regresión con el caso reproducido antes de re-abrir la tool
Post-mortem -> payload -> golden set (161); actualizar matriz de riesgo (169) y notificar afectados
```

El IR no borra el incidente: lo contiene con evidencia intacta y lo vuelve una defensa permanente.
