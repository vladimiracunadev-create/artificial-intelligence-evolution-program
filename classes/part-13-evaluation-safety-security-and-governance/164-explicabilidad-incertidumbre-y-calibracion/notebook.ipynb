## 📖 Resumen de la materia

Dos preguntas distintas: **explicabilidad** (¿por qué esta predicción?) e **incertidumbre/
calibración** (¿cuánto confiar?). **LIME** ajusta un modelo lineal local perturbando la entrada;
**SHAP** reparte la predicción entre features con valores de Shapley (propiedad de **aditividad**:
suman la diferencia con el valor base). Ambos explican el *modelo*, no la causa del mundo.

Incertidumbre **aleatoria** (ruido irreducible, no baja con datos) vs **epistémica** (ignorancia
del modelo, sí baja). Un modelo está **calibrado** si su confianza coincide con la frecuencia real:
`P(Y=1|score=p)=p`. Calibración ≠ exactitud. El **ECE** agrupa predicciones en bins y promedia,
ponderado por tamaño, la brecha |acc − conf| de cada bin. Recalibración típica: **temperature
scaling** (no cambia el ranking, solo las confianzas).


## 🧮 Mini-ejemplo: ECE con 3 bins

10 predicciones agrupadas por confianza:

```text
Bin        n   acc     conf    |acc-conf|
[0.5,0.7)  3   0.333   0.600   0.267
[0.7,0.9)  4   0.750   0.8125  0.0625
[0.9,1.0]  3   0.667   0.950   0.283
ECE = 0.3*0.267 + 0.4*0.0625 + 0.3*0.283 = 0.190
```

ECE ≈ 0.19: modelo **sobreconfiado**, especialmente en [0.9,1.0] (dice 0.95, acierta 0.67).
