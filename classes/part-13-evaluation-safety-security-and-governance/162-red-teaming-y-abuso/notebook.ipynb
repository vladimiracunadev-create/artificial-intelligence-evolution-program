## 📖 Resumen de la materia

**Red teaming** es la búsqueda adversarial, autorizada y sistemática de fallos: los casos se
construyen para romper, el éxito se mide en fallos encontrados y el entregable es un reporte con
severidad y reproducción. Requiere un **modelo de amenazas** (quién, qué quiere, qué puede, por
dónde) y una taxonomía de familias: elusión de salvaguardas, inyección de prompt, extracción,
**abuso** (uso dañino de la capacidad legítima), degradación y manipulación del modelo.

Métricas: **ASR** (attack success rate) *por familia y severidad* — nunca agregado único —,
cobertura de la matriz amenaza × superficie, y tiempo-hasta-fallo. Todo hallazgo confirmado se
convierte en ítem del golden set de regresión (clase 161). El red teaming demuestra presencia de
fallos, nunca su ausencia.


## 🧮 Mini-ejemplo

