## Reflexión (guía)

1. El **uso previsto** (contexto de aplicación), no la tecnología: el principio es que la misma
   capacidad técnica implica obligaciones distintas según el daño potencial de su aplicación.
2. El **desempeño por subgrupo** y los **usos no recomendados**: sin desglose, un número agregado
   oculta disparidades; sin límites de uso, la ficha no protege contra el mal uso ni permite auditar.
3. Porque la evidencia fabricada al final no es trazable ni reproducible; la cadena de evidencia
   exige que cada medición registre datos, protocolo, responsable y fecha a lo largo del ciclo.
