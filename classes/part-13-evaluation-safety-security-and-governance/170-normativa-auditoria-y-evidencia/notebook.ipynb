## 📖 Resumen de la materia

El **EU AI Act (Reglamento UE 2024/1689)** clasifica la IA por riesgo: **inaceptable** (prohibido),
**alto** (obligaciones estrictas: gestión de riesgos, gobernanza de datos, documentación técnica,
trazabilidad, transparencia, **supervisión humana**, exactitud/robustez/seguridad), **limitado**
(transparencia: avisar que es IA) y **mínimo** (sin obligaciones). Clave: el **uso**, no la
tecnología, determina el nivel.

Los artefactos de transparencia: **model cards** (Mitchell et al., arXiv:1810.03993) documentan un
modelo con desempeño *por subgrupo* y *usos no recomendados*; **datasheets** (Gebru et al.,
arXiv:1803.09010) documentan un dataset (motivación, composición, recolección, usos). Una
**auditoría** verifica el sistema contra un criterio y requiere una **cadena de evidencia**
reproducible producida durante todo el ciclo, no fabricada al final.
