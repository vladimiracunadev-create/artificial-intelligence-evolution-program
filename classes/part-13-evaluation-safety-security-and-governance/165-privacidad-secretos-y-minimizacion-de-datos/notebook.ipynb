## 📖 Resumen de la materia

Un modelo **memoriza** cuando reproduce literalmente secuencias de su corpus. Carlini et al.
(arXiv:2012.07805) mostraron el **ataque de extracción de datos de entrenamiento**: recuperar
cadenas memorizadas (PII, secretos) solo consultando el modelo. La memorización **crece con el
tamaño del modelo y la repetición** del dato; los datos únicos son los más identificables si se
extraen.

Los secretos se filtran en cuatro puntos: entrenamiento (memorización), contexto/RAG, logs de
prompts (la fuga más común y más fácil de evitar) y salida. La **minimización de datos** —recolectar,
usar y retener el mínimo necesario— es la defensa base. Técnicas: redacción, seudonimización,
tokenización de secretos, retención con TTL y **privacidad diferencial** (garantía formal vía
epsilon, con coste de utilidad). La fuga se mide con **canarios** (cadenas únicas plantadas para
auditar memorización).


## 🧮 Mini-ejemplo: canarios

```text

```

La memorización crece con la repetición → deduplicar el corpus es minimización efectiva. Y **tasa
baja ≠ riesgo bajo**: un canario 1x extraído expone a una persona única.
