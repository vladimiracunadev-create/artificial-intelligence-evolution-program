## 📖 Resumen de la materia

Una **alucinación** es una salida presentada como factual sin respaldo en la entrada, el contexto o
el mundo: **intrínseca** (contradice la fuente) o **extrínseca** (añade lo no verificable). Nace de
que el objetivo generativo premia lo plausible, no lo verdadero.

Dos palancas: **grounding** (recuperar, condicionar, atribuir, verificar — el motor del RAG) reduce
la alucinación extrínseca y hace la respuesta verificable; **abstención selectiva** (responder solo
si la confianza supera un umbral, si no "no sé"/derivar) reduce el error donde sí se responde. Se
mide con **cobertura** (fracción respondida) y **riesgo selectivo** (error entre lo respondido): la
**curva riesgo-cobertura** baja el riesgo al subir el umbral, *siempre que la confianza esté
calibrada* (clase 167). Reportar solo accuracy global oculta este trade-off.


## 🧮 Mini-ejemplo: riesgo-cobertura

QA médico, 20 respuestas, 10 correctas:

```text
Política      cobertura   riesgo (error entre respondidas)
sin abstener    1.00        0.500   <- responde todo: peor en dominio crítico
umbral 0.80     0.50        0.200
umbral 0.90     0.30        0.167
```

Subir el umbral baja cobertura y riesgo. La abstención solo funciona si la confianza está calibrada.
