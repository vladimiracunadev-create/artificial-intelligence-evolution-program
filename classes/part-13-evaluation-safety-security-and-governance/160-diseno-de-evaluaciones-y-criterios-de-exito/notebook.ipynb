## 📖 Resumen de la materia

Una **evaluación** es un procedimiento reproducible con cuatro componentes: datos (**D**), tarea y
protocolo (**T**), métrica (**M**) y criterio de éxito (**C**) declarado *antes* de medir. La
**validez de constructo** pregunta si la medida captura la capacidad que dice medir; sus amenazas
principales son la subrepresentación, la varianza irrelevante (longitud, formato), la
**contaminación de benchmarks** (ítems de test vistos en entrenamiento, detectable con ítems
perturbados) y la **ley de Goodhart** (optimizar la métrica degrada el constructo).

Un criterio de éxito útil tiene umbral, **baseline** y costo de error: 90 % de accuracy no
significa nada si predecir siempre la clase mayoritaria ya da 90 %.
