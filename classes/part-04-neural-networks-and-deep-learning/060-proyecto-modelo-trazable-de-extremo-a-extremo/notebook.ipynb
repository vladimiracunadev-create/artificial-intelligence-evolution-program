## 📖 Materia resumida

**Trazabilidad**: cada afirmación sobre un modelo debe seguirse hasta su origen —
datos (hash), splits (semilla), config, commit, checkpoint, protocolo de evaluación.
Un eslabón sin registrar convierte la métrica en anécdota.

**Documentación estándar**: *Datasheets for Datasets* (Gebru et al., 2018) para los
datos; *Model Cards* (Mitchell et al., 2019) para el modelo — uso previsto, métricas
desagregadas por subgrupo, condiciones de fallo, consideraciones éticas.

**Protocolo honesto**: métrica y baseline decididos *antes* de mirar el test; el test
se toca una vez; reportar media ± desviación sobre varias semillas; desagregar por
subgrupos. Reproducibilidad = semillas + entorno fijado + datos versionados (y saber
que el determinismo bit a bit en GPU tiene coste).
