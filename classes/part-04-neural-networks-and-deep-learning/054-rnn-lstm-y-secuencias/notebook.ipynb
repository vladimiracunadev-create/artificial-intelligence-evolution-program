## 📖 Materia resumida

**RNN**: `h_t = tanh(W_x·x_t + W_h·h_{t−1} + b)` con pesos compartidos en el tiempo;
se entrena desplegándola (BPTT). El gradiente entre pasos lejanos atraviesa k
productos de jacobianos: si su norma < 1 **desaparece** (∝ w^k), si > 1 **explota**
(esto último se recorta con gradient clipping; lo primero no tiene arreglo sin
cambiar la arquitectura — Bengio et al., 1994).

**LSTM** (Hochreiter y Schmidhuber, 1997): estado de celda con actualización aditiva
`c_t = f⊙c_{t−1} + i⊙c̃_t` y compuertas σ de olvido/entrada/salida. Con f ≈ 1 el
gradiente fluye casi intacto — el análogo temporal del atajo residual. GRU: versión
de 2 compuertas, similar con menos parámetros.

**Límites estructurales**: cómputo secuencial (no paralelizable en t) y memoria en un
vector fijo — exactamente lo que la atención (clase 055) elimina.
