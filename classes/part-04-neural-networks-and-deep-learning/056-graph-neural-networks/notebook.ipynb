## 📖 Materia resumida

**GNN**: deep learning sobre grafos G = (V, E) con el requisito de **invarianza a
permutaciones** de nodos. Primitiva universal — paso de mensajes (Gilmer et al., 2017):

```text
m_v = AGGREGATE({h_u : u ∈ vecinos(v)})   # suma/media/máx (sin orden)
h_v' = UPDATE(h_v, m_v)
```

k capas ⇒ cada nodo ve su vecindario a distancia ≤ k (el "campo receptivo" del grafo).

**GCN** (Kipf y Welling, 2016): `H' = σ(D̂^{-1/2}·Â·D̂^{-1/2}·H·W)` — media de vecinos
normalizada por grado + proyección compartida. **GAT**: pesos por vecino aprendidos con
atención (la clase 055 restringida a las aristas).

**Over-smoothing**: promediar repetidamente acerca todas las representaciones; las GNN
prácticas usan 2-4 capas. Límite expresivo: test de Weisfeiler-Leman.
