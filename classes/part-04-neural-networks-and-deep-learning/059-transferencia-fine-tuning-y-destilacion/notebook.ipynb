## 📖 Materia resumida

**Transferencia**: las capas tempranas aprenden features genéricas → se reutilizan.
*Feature extraction* (congelar tronco + cabeza nueva) con pocos datos; *fine-tuning*
(η 10-100× menor) con más datos — vigilando el olvido catastrófico.

**LoRA** (Hu et al., 2021): congela W y aprende ΔW = B·A de rango r ≪ dim →
`r·(d+k)` parámetros en vez de `d·k` (≈ 2 % con r=8 en 768×768); adaptadores por tarea
de MB, fusionables en W para servir sin sobrecoste.

**Destilación** (Hinton et al., 2015): el student imita el softmax del teacher con
temperatura T > 1 (`p_i = exp(z_i/T)/Σexp(z_j/T)`), que revela similitudes entre
clases; pérdida = λ·CE(etiquetas) + (1−λ)·T²·KL(teacher‖student).
