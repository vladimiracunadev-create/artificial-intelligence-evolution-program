## 📖 Materia resumida

**Q-learning**: `Q(s,a) ← Q(s,a) + α·[r + γ·max_a' Q(s',a') − Q(s,a)]` — aprende el
valor de acción óptimo off-policy. **DQN** (Mnih et al., 2015) lo lleva a redes
profundas con dos estabilizadores imprescindibles: *replay buffer* (rompe la
correlación temporal) y *target network* θ⁻ congelada (objetivo fijo por N pasos).
Exploración: ε-greedy.

**Policy gradient** (REINFORCE): `∇J = E[Σ ∇log π_θ(a|s)·G]` — sube la probabilidad
de acciones seguidas de retornos altos; alta varianza, se reduce restando una baseline
V(s) (actor-critic, ventaja A = G − V). **PPO** recorta el ratio de políticas para
evitar pasos destructivos: el estándar actual (robótica, juegos, RLHF).

**Trade-off**: valores = eficiencia de datos con acciones discretas; política = acciones
continuas y estocásticas con datos frescos.
