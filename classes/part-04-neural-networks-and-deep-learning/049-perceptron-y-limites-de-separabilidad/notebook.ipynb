## 📖 Materia resumida

**Perceptrón** (Rosenblatt, 1958): clasificador lineal `ŷ = escalón(w·x + b)`.
La frontera de decisión es el hiperplano `w·x + b = 0`.

**Regla de aprendizaje**: solo corrige ante error —
`w ← w + η(y − ŷ)x`, `b ← b + η(y − ŷ)`. El teorema de convergencia (Novikoff)
garantiza terminación en a lo sumo (R/γ)² actualizaciones **si** los datos son
linealmente separables con margen γ.

**Límite (Minsky y Papert, 1969)**: XOR no es linealmente separable — el sistema de
desigualdades `b<0`, `w₂+b≥0`, `w₁+b≥0`, `w₁+w₂+b<0` es contradictorio. Este resultado
motivó las capas ocultas y backpropagation (clase 050).
