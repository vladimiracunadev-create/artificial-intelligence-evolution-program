## 📖 Materia resumida

**SGD**: `θ ← θ − η·g` con g estimado sobre un minibatch. **Momento**: `v ← μv + g`,
`θ ← θ − ηv` — acelera direcciones consistentes, amortigua oscilaciones.

**Adam** (Kingma y Ba, 2014): medias móviles `m` (gradiente) y `v` (gradiente²),
corrección de sesgo `m̂ = m/(1−β₁ᵗ)`, `v̂ = v/(1−β₂ᵗ)`, paso `θ ← θ − α·m̂/(√v̂+ε)`.
Resultado: tasa efectiva por parámetro, pasos de magnitud ≈ α. **AdamW** separa el
weight decay de la parte adaptativa (estándar en Transformers).

**Regularización**: dropout (anular con prob. p y escalar por 1/(1−p); apagado en
inferencia), weight decay, early stopping. **Schedulers**: warmup lineal al inicio +
decaimiento (step o cosine). El primer paso de Adam vale exactamente −α·g/|g|:
compruébalo en el ejemplo trabajado del README.
