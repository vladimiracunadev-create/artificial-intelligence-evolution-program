## 📖 Materia resumida

**Autoencoder**: encoder-cuello de botella-decoder, pérdida de reconstrucción; el caso
lineal equivale a PCA. **VAE** (Kingma y Welling, 2013): latente probabilístico
`z = μ + σ·ε` (reparametrización) + ELBO = reconstrucción − KL(q‖N(0,I)) → latente
muestreable, generación real.

**GAN** (Goodfellow et al., 2014): juego minimax generador vs. discriminador;
muestras nítidas en 1 pasada, pero entrenamiento inestable y riesgo de mode collapse.

**Difusión** (Ho et al., 2020): forward fijo `x_t = √ᾱ_t·x₀ + √(1−ᾱ_t)·ε`; se aprende
a predecir ε (regresión estable); generar = denoisear desde ruido puro en T pasos.
Estado del arte en calidad y cobertura, caro en inferencia.
