## 📖 Materia resumida

**Activaciones**: la sigmoide satura (derivada ≤ 0.25 → gradiente desaparece en redes
profundas); ReLU `max(0,z)` mantiene derivada 1 en la zona activa (riesgo: neuronas
muertas); GELU es la variante suave estándar en Transformers.

**Inicialización**: conservar la varianza capa a capa. `Var(z) = n·Var(w)·Var(x)`, luego:
- Xavier/Glorot (tanh/sigmoide): `Var(w) = 2/(n_in + n_out)`
- He (ReLU): `Var(w) = 2/n_in` (el factor 2 compensa la mitad anulada por ReLU)

**Batch norm** (Ioffe y Szegedy, 2015): `x̂ = (x−μ_batch)/√(σ²_batch+ε)`, `y = γx̂ + β`.
Estabiliza el entrenamiento y regulariza; en inferencia usa medias móviles
(por eso `model.eval()` cambia el resultado). Layer norm normaliza por ejemplo,
no por batch: estándar en secuencias y Transformers.
