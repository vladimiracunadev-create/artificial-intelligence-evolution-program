## 📖 Materia resumida

**Convolución**: kernel compartido deslizado sobre la entrada —
`S(i,j) = Σₘ Σₙ I(i+m, j+n)·K(m,n) + b`. Conectividad local + pesos compartidos =
equivarianza a traslación y pocos parámetros (`C_out·(K²·C_in+1)`).

**Tamaño de salida**: `⌊(N − K + 2P)/S⌋ + 1`. Dos conv 3×3 apiladas ven un campo
receptivo 5×5 (con menos parámetros que un 7×7 directo — diseño VGG).

**Pooling**: reduce resolución con invarianza local; sin parámetros.

**Evolución**: LeNet-5 (1998, ~60k params) → AlexNet (2012, ReLU+GPU) → VGG (3×3) →
**ResNet** (2015): bloques `y = F(x) + x` cuyo atajo da al gradiente una vía directa
y permite 100+ capas (resuelve la degradación, no el sobreajuste).
