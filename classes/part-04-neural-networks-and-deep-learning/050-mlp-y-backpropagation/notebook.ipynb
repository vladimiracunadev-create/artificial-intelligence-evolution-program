## 📖 Materia resumida

**MLP**: composición de capas afines + no linealidades: `h = σ(W₁x + b₁)`,
`ŷ = W₂h + b₂`. Sin la no linealidad, todo colapsa en una sola capa lineal.
El teorema de aproximación universal garantiza expresividad, no entrenabilidad.

**Backpropagation** (Rumelhart et al., 1986): regla de la cadena organizada sobre el
grafo de cómputo, en dos pasadas:

1. *Forward*: evaluar guardando intermedios.
2. *Backward*: `δ_capa = (Wᵀ·δ_siguiente) ⊙ f'(z)`; `∂L/∂W = δ·hᵀ`; coste O(pesos).

**Actualización**: `θ ← θ − η·∂L/∂θ`. El paisaje no es convexo: no hay garantía de
mínimo global, pero el gradiente exacto y barato hace viable el entrenamiento profundo.
