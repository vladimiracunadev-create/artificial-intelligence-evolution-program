## 📖 Materia en 5 minutos

**Por qué probabilidad y no lógica.** Las reglas lógicas colapsan ante excepciones: `dolor → caries` es falsa (puede ser gingivitis) y su inversa también. La probabilidad resume pereza, ignorancia teórica e ignorancia práctica en un **grado de creencia**: `P(caries | dolor) = 0.6` no dice que la caries sea "60 % verdadera", sino cuánto creemos en ella dada la evidencia.

**Las tres operaciones sobre la distribución conjunta** `P(X₁,…,Xₙ)`:

1. **Marginalización:** `P(X) = Σ_y P(X, y)` — sumar fuera lo que no interesa.
2. **Condicionamiento:** `P(X | e) = P(X, e) / P(e)` — restringir a los mundos compatibles con la evidencia y renormalizar.
3. **Regla del producto:** `P(X, Y) = P(X | Y) · P(Y)` — encadenada da la regla de la cadena.

**Independencia condicional** `X ⊥ Y | Z`: `P(X, Y | Z) = P(X|Z)·P(Y|Z)`. Es el arma contra la explosión combinatoria: una conjunta de 32 variables binarias tiene ~4.3×10⁹ parámetros; con independencias condicionales pueden bastar decenas.
