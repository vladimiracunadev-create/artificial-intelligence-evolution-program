## 📖 Materia en 5 minutos

**Idea central.** Cuando una cantidad `E[f(X)]` no se puede calcular analíticamente, se **estima muestreando**: genera `x₁,…,x_N ~ P`, promedia `f`:

```text
Ê = (1/N) Σ f(xᵢ)
```

- **Ley de los grandes números:** `Ê → E[f(X)]` cuando `N → ∞` (el estimador es insesgado y consistente).
- **Teorema central del límite:** el error estándar decrece como `σ/√N`. Regla práctica: **para dividir el error entre 2 hacen falta ×4 muestras**; un dígito decimal más cuesta ×100.

**Variantes que verás en el curso:** muestreo por rechazo (descarta muestras incompatibles con la evidencia — se desperdicia trabajo si la evidencia es rara), muestreo por importancia (pondera muestras de una distribución auxiliar) y MCMC (cadenas que convergen a la distribución objetivo).

**Reproducibilidad.** Un experimento Monte Carlo sin semilla declarada no es evidencia: dos corridas legítimas pueden diferir. Reportar `Ê ± error estándar` y la semilla convierte una simulación en un resultado auditable.
