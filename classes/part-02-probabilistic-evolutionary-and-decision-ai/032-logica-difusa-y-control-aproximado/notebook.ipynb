## 📖 Materia en 5 minutos

**Conjunto difuso** = conjunto con **función de membresía** `μ_A: X → [0,1]`. `μ_alto(1.78) = 0.6` significa que 1.78 m pertenece "en grado 0.6" al predicado *alto*. Ojo: modela **vaguedad del lenguaje**, no incertidumbre sobre hechos — es una cantidad distinta de una probabilidad.

**Operadores estándar (Zadeh):**

```text
NOT:  μ_¬A(x) = 1 − μ_A(x)
AND:  μ_A∧B(x) = min(μ_A(x), μ_B(x))
OR:   μ_A∨B(x) = max(μ_A(x), μ_B(x))
```

**Controlador difuso (Mamdani), 4 etapas:**

1. **Fuzzificación:** el valor nítido de entrada se convierte en grados de membresía.
2. **Evaluación de reglas:** `SI temperatura es alta Y humedad es baja ENTONCES ventilador rápido` — el antecedente se combina con `min`, y su grado **recorta** el conjunto de salida.
3. **Agregación:** los conjuntos recortados de todas las reglas se unen con `max`.
4. **Defuzzificación:** el conjunto agregado se colapsa a un número, típicamente por **centroide** `∫x·μ(x)dx / ∫μ(x)dx`.

La gracia práctica: codifica conocimiento experto lingüístico con transiciones suaves entre reglas, sin modelo matemático del proceso.
