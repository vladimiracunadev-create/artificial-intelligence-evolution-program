## Soluciones explicadas

**E1.** `EU(con) = 20`. `EU(sin) = 0.4·(−80) + 0.6·100 = −32 + 60 = 28`. Se decide **sin paraguas**: la probabilidad de lluvia no basta para compensar los 100 del día libre de estorbo.

**E2.** `20 = p·(−80) + (1−p)·100` → `20 = 100 − 180p` → `p = 4/9 ≈ 0.444`. Sobre ese umbral conviene el paraguas. Nota lo sensible que es la decisión: el `p=0.4` de E1 está a solo 0.044 del cruce — un análisis de sensibilidad honesto lo declararía frágil.

**E3.** `EU(lotería) = 0.5·√0 + 0.5·√10000 = 50`. `U(4000) = √4000 ≈ 63.2`. Se prefieren los 4 000 seguros. Equivalente cierto: `U(EC) = 50` → `EC = 2500`. Prima de riesgo `= 5000 − 2500 = 2500`: el agente sacrifica la mitad del valor esperado por eliminar la varianza.

**E4.** Con pronóstico perfecto se elige lo mejor en cada rama: `EU = 0.4·max(20, −80) + 0.6·max(20, 100) = 0.4·20 + 0.6·100 = 68`. `VOI = 68 − 28 = 40 > 10` → sí, se paga. El VOI es alto justamente porque la decisión de E1 estaba cerca del punto de indiferencia.

```python
import math
p = 0.4
eu_con = 20
eu_sin = p*(-80) + (1-p)*100
p_star = 80/180  # de 100-180p = 20
eu_lot = 0.5*math.sqrt(0) + 0.5*math.sqrt(10000)
ec = eu_lot**2
eu_perf = p*max(20, -80) + (1-p)*max(20, 100)
print(f"E1 con={eu_con} sin={eu_sin} | E2 p*={p_star:.3f}")
print(f"E3 EU_lot={eu_lot} EC={ec:.0f} prima={5000-ec:.0f} | E4 VOI={eu_perf-eu_sin:.0f}")
```
