## 📖 Materia en 5 minutos

**Decisión racional = probabilidad + preferencias.** Una **lotería** es una distribución sobre resultados. Los axiomas de von Neumann-Morgenstern (orden, transitividad, continuidad, independencia) implican que existe una función de **utilidad** `U` tal que el agente racional elige la acción de **máxima utilidad esperada**:

```text
a* = argmax_a  EU(a) = argmax_a  Σ_s P(s | a, e) · U(resultado(a, s))
```

**Utilidad ≠ dinero.** La utilidad del dinero es típicamente cóncava (aversión al riesgo): `U(√·)` o `U(log·)` hacen que una lotería 50/50 entre 0 y 100 valga *menos* que 50 seguros. El **equivalente cierto** es la cantidad segura que iguala la utilidad esperada de la lotería.

**Valor de la información (VOI).** Cuánto vale observar una variable antes de decidir: `VOI = EU(con la observación, decidiendo después) − EU(decidiendo ya)`. Nunca es negativo para un agente racional y es 0 si la observación no puede cambiar la decisión.
