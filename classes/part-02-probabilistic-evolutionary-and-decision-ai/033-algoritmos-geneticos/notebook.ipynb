## 📖 Materia en 5 minutos

**Algoritmo genético (GA)** = búsqueda poblacional inspirada en la selección natural. No necesita gradientes ni continuidad: solo una función de **fitness** que puntúe candidatos.

```text
1. Inicializar población de N individuos (p. ej., cadenas binarias)
2. Repetir por generaciones:
   a. Evaluar fitness de cada individuo
   b. SELECCIÓN   : elegir padres sesgados hacia mayor fitness
                    (ruleta: p_i = f_i / Σf ; o torneo de tamaño k)
   c. CRUCE       : combinar pares de padres (un punto: cortar y pegar colas)
   d. MUTACIÓN    : voltear cada bit con probabilidad p_m pequeña
   e. REEMPLAZO   : formar la nueva generación (elitismo: conservar al mejor)
3. Devolver el mejor individuo visto
```

**El balance central — exploración vs explotación:** la selección explota lo bueno conocido; cruce y mutación exploran lo nuevo. Presión de selección alta + mutación baja → convergencia prematura a un óptimo local; lo contrario → deriva aleatoria sin progreso.

**Elitismo:** copiar el mejor individuo intacto garantiza que el fitness máximo nunca retrocede entre generaciones.
