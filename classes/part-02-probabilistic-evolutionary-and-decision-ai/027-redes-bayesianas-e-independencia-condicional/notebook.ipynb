## 📖 Materia en 5 minutos

**Red bayesiana** = grafo dirigido acíclico (DAG) donde cada nodo `Xᵢ` lleva una tabla de probabilidad condicional (CPT) `P(Xᵢ | padres(Xᵢ))`. La red representa la conjunta completa de forma compacta:

```text
P(X₁,…,Xₙ) = Π_i P(Xᵢ | padres(Xᵢ))
```

Con `n` variables binarias y a lo sumo `k` padres, los parámetros pasan de `2ⁿ−1` a `≤ n·2ᵏ`.

**Las tres conexiones básicas** (claves para leer independencias):

- **Cadena** `A → B → C`: A y C dependientes; **independientes dado B** (B bloquea).
- **Causa común (fork)** `A ← B → C`: A y C dependientes; **independientes dado B**.
- **Colisionador (collider)** `A → B ← C`: A y C **independientes a priori**; observar B (o un descendiente) los **conecta** — el efecto *explaining away*: si suena la alarma y confirmo robo, baja mi creencia en el terremoto.

**d-separación.** `X ⊥ Y | Z` si *todo* camino no dirigido entre X e Y está bloqueado: un nodo de cadena o fork que esté en Z, o un colisionador que NO esté en Z (ni sus descendientes).
