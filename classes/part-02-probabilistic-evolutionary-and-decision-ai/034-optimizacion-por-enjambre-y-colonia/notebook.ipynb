## 📖 Materia en 5 minutos

**Inteligencia de enjambre:** muchos agentes simples + comunicación indirecta = búsqueda colectiva sin controlador central. Dos algoritmos canónicos:

**PSO (Particle Swarm Optimization).** Cada partícula `i` tiene posición `xᵢ` y velocidad `vᵢ`, recuerda su mejor posición personal `pᵢ` (pbest) y conoce la mejor global `g` (gbest):

```text
vᵢ ← w·vᵢ + c₁·r₁·(pᵢ − xᵢ) + c₂·r₂·(g − xᵢ)
xᵢ ← xᵢ + vᵢ
     w  : inercia (explorar)     c₁ : componente cognitiva (mi memoria)
     c₂ : componente social (el grupo)     r₁, r₂ ~ U(0,1)
```

**ACO (Ant Colony Optimization).** Hormigas construyen soluciones (p. ej., rutas) eligiendo aristas con probabilidad proporcional a `feromona^α · heurística^β`; al final se deposita feromona en las buenas rutas y **toda** la feromona se evapora:

```text
τ ← (1−ρ)·τ + Σ Δτ      (ρ: tasa de evaporación; Δτ ∝ 1/costo de la ruta)
```

La evaporación es el mecanismo de olvido: sin ella, las primeras rutas buenas monopolizan la búsqueda (estancamiento). PSO opera en espacios continuos; ACO brilla en combinatorios (grafos).


### Mini ejemplo

Una partícula en 1D: `x = 4`, `v = 1`, `pbest = 3`, `gbest = 0`, con `w = 0.5`, `c₁ = c₂ = 1` y `r₁ = r₂ = 0.5`:
`v' = 0.5·1 + 0.5·(3−4) + 0.5·(0−4) = 0.5 − 0.5 − 2 = −2` → `x' = 4 − 2 = 2`. La partícula es atraída hacia el consenso del enjambre. El laboratorio `optimization` reporta la traza de mejora con semilla fija.
