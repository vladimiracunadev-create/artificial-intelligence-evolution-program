## Ejercicios

MDP de 2 estados `s∈{0,1}` y 2 acciones. Desde `s=0`: acción `a` lleva a `s=1` con prob. 0.8 y recompensa +10 (si falla, se queda en 0 con recompensa 0); acción `b` se queda en 0 con recompensa +1 seguro. El estado `s=1` es terminal con `V(1)=0`. Usa `γ = 0.8`.

**E1 — Un paso de value iteration a mano.** Con `V₀ = (0, 0)`, calcula `Q(0,a)`, `Q(0,b)` y `V₁(0)`. ¿Qué acción prefiere la política?

**E2 — Segundo paso.** Calcula `V₂(0)` usando `V₁`. Observa el patrón: ¿hacia qué valor converge `V(0)`? (pista: resuelve `v = 0.8·10 + 0.2·γ·v` y verifica que en ese punto `Q(0,b) < v`).

**E3 — El descuento decide.** Los puntos fijos "puros" son `v_a = 8/(1−0.2γ)` (siempre a) y `v_b = 1/(1−γ)` (siempre b). ¿Para qué valor de γ empatan? Interpreta: ¿qué agente (miope/paciente) prefiere `b` y por qué?

**E4 — Recompensa con signo.** Si cada intento de `a` costara −2 (recompensa 0.8·10 − 2 en esperanza), recalcula E1 y decide si la política cambia.


```python
# TU RESPUESTA AQUÍ
gamma = 0.8

def q_a(v0):  # Q(0, a)
    return 0.8*(10 + gamma*0) + 0.2*(0 + gamma*v0)

def q_b(v0):  # Q(0, b)
    return 1 + gamma*v0

# E1: V1 = max(q_a(0), q_b(0))
# E2: itera; punto fijo v = 8/(1-0.2*gamma)
# E3: encuentra gamma de empate entre 8/(1-0.2g) y 1/(1-g)
```
