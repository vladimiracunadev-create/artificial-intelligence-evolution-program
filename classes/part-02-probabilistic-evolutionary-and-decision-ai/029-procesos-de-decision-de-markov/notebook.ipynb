## 📖 Materia en 5 minutos

**MDP** `(S, A, P, R, γ)`: estados, acciones, transiciones estocásticas `P(s'|s,a)`, recompensa `R(s,a,s')` y descuento `γ ∈ [0,1)`. La solución no es un plan fijo sino una **política** `π: S → A` — qué hacer en *cada* estado, porque el resultado de las acciones es incierto.

**Ecuación de optimalidad de Bellman:**

```text
V*(s) = max_a Σ_s' P(s'|s,a) · [ R(s,a,s') + γ · V*(s') ]
```

El valor óptimo de un estado es el mejor "promedio ponderado" de recompensa inmediata más futuro descontado.

**Value iteration:** inicializa `V₀ = 0` y aplica la ecuación como operador de actualización hasta converger (es una contracción con factor γ: la convergencia está garantizada). La política óptima se extrae al final con `argmax_a`.

**Papel de γ:** cercano a 0 → agente miope (solo recompensa inmediata); cercano a 1 → planificador de largo plazo, pero más iteraciones para converger.


### Mini ejemplo

Dos estados `{frío, caliente}` y acciones `{esperar, actuar}`. Si en `frío` actúas, pasas a `caliente` con prob. 0.8 (recompensa 10) o te quedas (recompensa 0). Un paso de value iteration desde `V₀=(0,0)`:
`Q(frío, actuar) = 0.8·(10 + γ·0) + 0.2·(0 + γ·0) = 8`. Con `Q(frío, esperar) = 1`, la política elige `actuar`. El laboratorio `workflow` encadena decisiones de este tipo con trazabilidad de cada paso.
