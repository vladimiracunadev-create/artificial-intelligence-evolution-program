## Soluciones explicadas

**E1.** `P(+) = 0.85·0.02 + 0.10·0.98 = 0.017 + 0.098 = 0.115`. Posterior `= 0.017/0.115 ≈ 0.148`. Aun con un test decente, un positivo aislado en población de baja prevalencia deja la creencia bajo 15 %.

**E2.** Nuevo prior 0.148: `P(+) = 0.85·0.148 + 0.10·0.852 = 0.1258 + 0.0852 = 0.211`; posterior `= 0.1258/0.211 ≈ 0.596`. Dos positivos independientes sí superan el 50 %: la evidencia se acumula multiplicativamente en odds (LR² = 72.25).

**E3.** Odds prior `= 0.0204`; `LR = 8.5`; odds posterior `= 0.1735`; probabilidad `= 0.1735/1.1735 ≈ 0.148`. ✔ Coincide — la forma odds es el mismo teorema sin calcular `P(e)`.

**E4.** `p/(1−p) = 1/LR = 1/8.5` → `p ≈ 0.105`. Solo si la prevalencia supera ~10.5 % un único positivo hace más probable estar enfermo que sano: el cribado masivo en población de bajo riesgo produce mayoría de falsos positivos por diseño.


```python
result = run_lab("probability", seed=26)
assert result["kind"] == "probability"
assert result["evidence"]
show(result)
```

```python
prior, sens, esp = 0.02, 0.85, 0.90
p1 = sens*prior / (sens*prior + (1-esp)*(1-prior))
p2 = sens*p1 / (sens*p1 + (1-esp)*(1-p1))
LR = sens/(1-esp)
odds = prior/(1-prior) * LR
p_odds = odds/(1+odds)
p_eq = 1/(1+LR)
print(f"E1 {p1:.3f} | E2 {p2:.3f} | E3 {p_odds:.3f} | E4 prior de equilibrio {p_eq:.3f}")
assert abs(p1 - p_odds) < 1e-12
```
