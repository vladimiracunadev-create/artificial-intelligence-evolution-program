## 📖 Materia en 5 minutos

**El teorema.** De la regla del producto `P(H,e) = P(e|H)P(H) = P(H|e)P(e)` se despeja:

```text
P(H | e) = P(e | H) · P(H) / P(e)
posterior = verosimilitud · prior / evidencia
```

- **Prior** `P(H)`: creencia antes de observar.
- **Verosimilitud** `P(e|H)`: qué tan compatible es la evidencia con cada hipótesis.
- **Evidencia** `P(e) = Σ_h P(e|h)P(h)`: normalizador (teorema de probabilidad total).

**Forma odds** (evita calcular `P(e)`): `odds posterior = odds prior × razón de verosimilitud`, con `LR = P(e|H)/P(e|¬H)`.

**Actualización secuencial.** Con evidencias condicionalmente independientes dado `H`, el posterior de hoy es el prior de mañana: `P(H|e₁,e₂) ∝ P(e₂|H)·P(e₁|H)·P(H)`. El orden no importa; el resultado es el mismo.


### Mini ejemplo (el clásico que la intuición falla)

Enfermedad con prevalencia 1 %, test con sensibilidad 90 % y especificidad 95 %:

```text
P(E | +) = 0.9·0.01 / (0.9·0.01 + 0.05·0.99)
        = 0.009 / 0.0585 ≈ 0.154
```

Un positivo deja la probabilidad en ~15 %, no en 90 %: con priors bajos, los falsos positivos del 99 % sano dominan. Ignorar el prior es la *falacia de la tasa base*.
