## 4. Intuición

Una población de soluciones que se reproducen: las mejores dejan más descendencia, los hijos mezclan trozos de sus padres y de vez en cuando algo muta. No hay gradiente, no hay modelo del problema. Holland demuestra que ese proceso reparte los ensayos casi como conviene.


## 5. Concepto mínimo

```text
Bucle: evaluar → seleccionar → cruzar → mutar → repetir

Un ESQUEMA es un patrón con comodines: 11****…
La población evalúa implícitamente MUCHOS esquemas a la vez (paralelismo implícito),
y la reproducción proporcional a la aptitud les asigna ensayos de forma
cercana a la óptima del problema del bandido de dos brazos.
```
