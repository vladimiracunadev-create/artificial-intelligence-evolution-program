## 4. Intuición

Un detector con 99 % de sensibilidad y 99 % de especificidad. En un flujo donde uno de cada 10 000 eventos es un ataque, produce **102 alertas por cada ataque real**.


## 5. Concepto mínimo

```text
precisión = VP / (VP + FP)

Con clase base p muy pequeña, FP ≈ N·(1−especificidad) DOMINA a VP ≈ N·p·sensibilidad
        → la precisión colapsa aunque el modelo sea excelente
```

## 9. Salida interpretable

## 10. Comentario pedagógico

Esa exigencia no se formula como «un modelo mejor» sino como «una tasa de falsos positivos absurdamente baja», y esa reformulación es lo útil del artículo. Las otras cuatro razones —coste asimétrico, ausencia de datos de ataques nuevos, adversario que se adapta, y necesidad de explicar la alerta— no se arreglan con más datos ni con más parámetros.
