## 4. Intuición

Dos aspiradoras en un mundo de dos casillas. Una reacciona a lo que ve; la otra recuerda dónde ha estado. Con una medida de desempeño las dos empatan. Con otra, una gana y la otra pierde. Nada cambió en los agentes: cambió lo que decidimos contar.


## 5. Concepto mínimo

```text
agente = f(secuencia de percepciones) → acción

racional ≠ omnisciente ≠ perfecto
racional = maximiza la MEDIDA DE DESEMPEÑO esperada,
           dado lo que ha percibido y lo que sabe

El diseño se especifica con cuatro cosas: medida, entorno, actuadores y sensores.
```

## 11. Error o anti-patrón deliberado

Anti-patrón: evaluar «lo que el agente hace» en lugar de «lo que consigue en su entorno».


````python
print('Aspirar mucho no es limpiar bien: un agente que ensucia y limpia puntua alto por accion.')
print('La medida tiene que estar sobre el ESTADO DEL ENTORNO, no sobre la actividad del agente.')
print('Es el error clasico y el mas caro de descubrir tarde.')
````
