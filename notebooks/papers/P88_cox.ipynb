## 4. Intuición

¿Por qué probabilidad y no otra cosa? Cox no la defiende por tradición ni por frecuencias: pide tres condiciones mínimas a cualquier medida razonable de creencia y demuestra que solo hay una familia que las cumple. La probabilidad no se elige: se deduce.


## 5. Concepto mínimo

```text
Desiderata:
  1. los grados de creencia se representan con números reales
  2. en los casos extremos, las reglas coinciden con la lógica
  3. si una conclusión se alcanza por dos caminos, ambos dan el mismo grado

Teorema: cualquier medida que los cumpla es isomorfa a la probabilidad,
         es decir satisface la regla de la suma y la del producto.
```
