## 10. Comentario pedagógico

El checklist no garantiza que un resultado sea cierto: garantiza que sea **comprobable**. Son dos cosas distintas y confundirlas lleva a la decepción con los checklists. Lo que cambia es que el coste de auditar baja lo suficiente como para que alguien lo haga.
