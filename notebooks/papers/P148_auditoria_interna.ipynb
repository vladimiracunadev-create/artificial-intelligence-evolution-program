## 4. Intuición

«Faltan las etiquetas de subgrupo» es barato al recoger los datos e **imposible** cuando el sistema ya está construido. Auditar al final es auditar cuando ya no se puede corregir.


## 5. Concepto mínimo

```text
alcance → correspondencia → recogida de artefactos → pruebas → reflexión

cada etapa produce un ARTEFACTO, no una opinión
```

## 10. Comentario pedagógico

Ese reconocimiento explícito es lo que hace serio al artículo. La auditoría interna **no sustituye** a la externa ni a la regulación: no da cuenta pública, no tiene poder de veto y cubre lo que la organización decide mirar. Lo que sí hace es dejar los artefactos —hojas de datos, tarjetas de modelo, resultados desagregados— sin los cuales un auditor externo solo puede preguntar.
