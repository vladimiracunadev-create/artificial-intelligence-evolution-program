## 4. Intuición

Un modelo se publica con una exactitud del 91 %. Desagregada por subgrupo, va del 94 % al 58 %. El grupo peor servido es el 2 % de la muestra, así que no mueve la cifra global — y por eso nadie lo ve si nadie desagrega.


## 5. Concepto mínimo

```text
  métricas · datos de evaluación · datos de entrenamiento
  ANÁLISIS CUANTITATIVO DESAGREGADO · consideraciones éticas · advertencias
```

## 9. Salida interpretable

La global es **0,9118**. Por subgrupo va de **0,94** a **0,5778**: una brecha de **0,362**. Y el grupo peor servido son **180 casos de 8 300** — el 2,2 % de la muestra —, así que su mal resultado se diluye por completo en el agregado.


## 10. Comentario pedagógico

La sección clave de una tarjeta es el análisis cuantitativo **desagregado**, y es la que obliga a publicar el número por subgrupo en vez de la media. Las otras nueve secciones responden preguntas que normalmente se hacen tarde: para qué sirve, para qué **no**, con qué datos se evaluó y qué advertencias tiene.

## 12. Corrección

Lo que la desagregación hace visible:


````python
r = run_paper_lab('tarjetas_de_modelo', seed=7)['result']
print('exactitud global:', r['exactitud_global'])
for s in r['por_subgrupo']:
    print(f"  grupo {s['grupo']}  n={s['n']:>5}  exactitud={s['exactitud']}")
print('brecha:', r['brecha'])
````
