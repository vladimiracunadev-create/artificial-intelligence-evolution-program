## 4. Intuición

## 9. Salida interpretable

**5 lecturas incompatibles** para la misma cadena. Con 30 agentes y 10 lenguajes de contenido, **8 700** adaptadores punto a punto frente a **40** con capa común: **217,5× menos**. La performativa es una etiqueta prestada de la filosofía del lenguaje.


## 10. Comentario pedagógico

Las tres capas son la parte que envejece bien. Permiten cambiar el lenguaje del contenido sin tocar el transporte, y el transporte sin tocar la semántica. Treinta años después, el Model Context Protocol resuelve el mismo problema con la misma forma — y con la misma dificultad pendiente: KQML nunca fijó una semántica formal de sus performativas, y dos implementaciones podían entender «tell» de forma distinta.
