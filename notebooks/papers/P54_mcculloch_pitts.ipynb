```text
salida = 1  si  Σ wᵢ·xᵢ ≥ θ     y  ninguna entrada inhibitoria está activa

    AND  : w = (1, 1),  θ = 2
    OR   : w = (1, 1),  θ = 1
    NAND : w = (−1, −1), θ = −1
```

## 10. Comentario pedagógico

Esta ficha y la del perceptrón se confunden constantemente. McCulloch y Pitts responden «¿qué puede CALCULAR una red de neuronas?»; Rosenblatt, quince años después en [P01](../../papers/foundational/P01_perceptron/README.md), responde «¿puede APRENDER sus propios pesos?». Son dos preguntas distintas y el orden importa.
