## 4. Intuición

Turing no responde «¿pueden pensar las máquinas?». Declara que la pregunta está mal planteada y propone otra que sí se puede ejecutar: si un interrogador conversa por escrito con una persona y con una máquina y no las distingue, ¿qué queda de la pregunta original?


## 5. Concepto mínimo

```text
Juego de imitación:
    interrogador  ──preguntas escritas──▶  A (máquina)  y  B (persona)
    objetivo del interrogador: decidir cuál es cuál

No hay fórmula. Hay un PROTOCOLO, y su poder depende de las preguntas que se hagan.
```
