## 4. Intuición

La conferencia del premio Turing de 1976 resume veinte años en dos frases. La primera: pensar es manipular símbolos. La segunda: resolver problemas es buscar en un espacio de estados, y lo que hace viable esa búsqueda no es la potencia de cálculo sino saber **dónde no mirar**.


## 5. Concepto mínimo

```text
Hipótesis del sistema de símbolos físicos:
    símbolos + estructuras + procesos = medios necesarios y suficientes
    para la acción inteligente general

Hipótesis de la búsqueda heurística:
    resolver = generar y probar caminos en un espacio de estados,
    guiados por información sobre el problema

Coste de la búsqueda ciega: b^d      ← el muro
```
