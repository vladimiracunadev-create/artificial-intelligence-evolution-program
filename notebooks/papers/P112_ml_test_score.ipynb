## 4. Intuición

Dos equipos con la misma exactitud. Uno tiene pruebas de datos, de modelo, de infraestructura y de monitorización; el otro no tiene ninguna. La métrica no los distingue, y son sistemas completamente distintos.


## 5. Concepto mínimo

```text
Cuatro categorías × siete pruebas cada una:
    datos · modelo · infraestructura · monitorización

Puntuación = MÍNIMO entre categorías (no la suma)
    un sistema es tan robusto como su categoría más débil
```

## 10. Comentario pedagógico

El mínimo entre categorías no es una elección estética. Un sistema con datos impecables y sin capacidad de revertir no es robusto: es uno con un punto único de fallo bien documentado. Sumar permitiría compensar la ausencia total de monitorización con pruebas de datos excelentes, y eso no es cómo fallan los sistemas.
