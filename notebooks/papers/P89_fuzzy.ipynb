## 4. Intuición

«¿Hace calor?» no es una pregunta con respuesta sí o no, y tampoco es una pregunta sobre probabilidad: a 25 °C no hay un 70 % de probabilidad de que haga calor, hace calor **en un grado de 0,7**. Zadeh formaliza esa diferencia y construye un álgebra sobre ella.


## 5. Concepto mínimo

```text
Conjunto clásico:  μ_A(x) ∈ {0, 1}
Conjunto difuso :  μ_A(x) ∈ [0, 1]

Operadores de Zadeh:
    A ∩ B  →  min(μ_A, μ_B)        A ∪ B  →  max(μ_A, μ_B)        ¬A  →  1 − μ_A

La suma de pertenencias NO tiene por qué ser 1: no es una distribución.
```

## 9. Salida interpretable

A 25 °C la temperatura pertenece a la vez a «templado» y a «caluroso» con grados parciales. Las pertenencias **no suman 1** —en la tabla dan 1,0, 0,65, 1,0, 0,786 y 1,0—, y por eso un grado de pertenencia no es una probabilidad. El AND de Zadeh es el mínimo y es idempotente: min(0,7 · 0,7) = 0,7, mientras el producto da 0,49.
