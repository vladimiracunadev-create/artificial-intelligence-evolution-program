## 4. Intuición

Ocho puntos y una recta que los separa. Y otra. Y otra. Todas aciertan el 100 % en entrenamiento, así que la exactitud no sirve para elegir. Vapnik propone otro criterio: quedarse con la que pasa más lejos de todos los puntos.


## 5. Concepto mínimo

```text
Margen geométrico:  γ = mín_i  yᵢ(w·xᵢ + b) / ‖w‖

Problema:  maximizar γ  ⟺  minimizar ‖w‖²/2  sujeto a  yᵢ(w·xᵢ + b) ≥ 1

Solo los puntos con yᵢ(w·xᵢ + b) = 1 definen la solución: los VECTORES SOPORTE.
```

## 10. Comentario pedagógico

De ahí sale una propiedad muy útil: el modelo no depende del tamaño del conjunto sino del número de vectores soporte. Y una idea que reaparece en todas partes: cuando varias hipótesis explican los datos igual de bien, hace falta un criterio adicional. Aquí es el margen; en [P77](../../papers/foundational/P77_lasso/README.md) será la parsimonia.


## 11. Error o anti-patrón deliberado

Anti-patrón: elegir entre modelos que empatan en entrenamiento mirando el entrenamiento.


````python
r = run_paper_lab('svm', seed=7)['result']
print('separadores perfectos en entrenamiento:', r['separadores_validos_probados'])
print('margen del mejor :', r['mejor_por_margen']['margen'])
print('margen del peor  :', r['peor_separador_valido']['margen'])
print('Los dos aciertan 100%. La exactitud no los distingue.')
````

## 12. Corrección

El criterio que sí los distingue, y su justificación:


````python
r = run_paper_lab('svm', seed=7)['result']
print('vectores soporte:', r['puntos_que_definen_la_frontera'])
for v in r['vectores_soporte']:
    print('  ', v)
print('El resto de puntos podria moverse sin cambiar la frontera.')
````
