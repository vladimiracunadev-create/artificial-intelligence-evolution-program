## 4. Intuición

## 5. Concepto mínimo

```text
Denegar por defecto : accesible ⟺ hay un permiso explícito
Permitir por defecto: accesible ⟺ NO hay una denegación explícita

El error de omisión abre en un caso y cierra en el otro
```
