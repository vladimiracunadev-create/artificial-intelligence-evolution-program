## 4. Intuición

El coordinador no sabe quién es bueno en qué. En vez de mantener un registro que se desactualiza, anuncia la tarea y deja que cada nodo declare su coste — porque cada nodo sí lo sabe.


## 5. Concepto mínimo

```text
1. ANUNCIO     coordinador → todos:  «hay esta tarea»
2. OFERTAS     cada nodo → coordinador: «me cuesta X»
3. ADJUDICA    coordinador → ganador:   «es tuya»

nadie mantiene una lista de capacidades
```

## 9. Salida interpretable

Reparto ciego: la última tarea termina en **8,23**. Red de contratos: **5,05**, un **1,63×** mejor, y el desequilibrio de carga baja de 5,57 a 3,15. El precio son **234 mensajes** frente a 18 — **13× más** comunicación.


## 10. Comentario pedagógico

Esa razón de 13× es la que decide si el protocolo compensa. Si los nodos son homogéneos y sus capacidades conocidas, un planificador central lo hace mejor y más barato. La red de contratos gana exactamente cuando **nadie tiene el registro**, que es el caso de un sistema abierto donde los nodos entran y salen.
